## Ejercicio 1.2 — Pokémon y la PokéAPI

La [PokéAPI](https://pokeapi.co) es una API pública y gratuita con información de todos los Pokémon. Es un ejemplo perfecto de fuente de datos **externa y semiestructurada** que se consume mediante peticiones HTTP.

**Parte A — Exploración:**  
El código de abajo descarga la ficha del Pokémon número 25 (Pikachu). Ejecútalo y responde las preguntas que hay debajo.


````python
import requests

url = 'https://pokeapi.co/api/v2/pokemon/25'
respuesta = requests.get(url)
datos = respuesta.json()

# Mostramos solo algunas claves para no saturar la salida
print('Claves del JSON:', list(datos.keys()))
print('Nombre:',  datos['name'])
print('Peso (hectogramos):', datos['weight'])
print('Altura (decimetros):', datos['height'])
print('Tipos:', [t['type']['name'] for t in datos['types']])
print('Habilidades:', [h['ability']['name'] for h in datos['abilities']])
print('Estadisticas:')
for stat in datos['stats']:
    print(f"  {stat['stat']['name']}: {stat['base_stat']}")
````

**Preguntas sobre la Parte A:**

1. ¿Por qué decimos que este JSON es datos *semiestructurados* y no estructurados?
2. ¿Qué deberías hacer para convertir estos datos en un DataFrame de pandas con una fila por Pokémon y columnas como `nombre`, `peso`, `altura`, `tipo_1`, `hp`, `ataque`?

*Responde aquí:*

1. El formato JSON es legible y flexible, pero no directamente tabular
    
2. Estructurarlo de forma que aparezcan en columnas y filas utilizando un dataFrame y pasandole el conjunto de datos: pd.DataFrame(datos)


**Parte B — A programar:**  
Descarga los datos de **5 Pokémon** de tu elección (puedes usar su número o su nombre) y construye un DataFrame de pandas con al menos las columnas: `nombre`, `peso`, `altura`, `hp`, `ataque`, `defensa`, `tipo_1`.
Muestra el DataFrame al final.


````python
import pandas as pd
import requests


pokemon_ids = range(1,100)
registros = []

for pid in pokemon_ids:
    respuesta = requests.get(f'https://pokeapi.co/api/v2/pokemon/{pid}')
    respuesta.raise_for_status()
    pokemon = respuesta.json()

    # Las estadísticas vienen dentro de una lista anidada.
    stats = {
        stat['stat']['name']: stat['base_stat']
        for stat in pokemon['stats']
    }

    registros.append({
        'nombre': pokemon['name'],
        'peso': pokemon['weight'],       # busca la clave correspondiente en pokemon
        'altura': pokemon['height'],
        'hp': stats['hp'],
        'ataque': stats['attack'],
        'defensa': stats['defense'],
        'tipo_1': pokemon['types'][0]['type']['name']
    })

df_pokemon = pd.DataFrame(registros)
print(df_pokemon)
````

## Ejercicio 2.4 — XML del mundo real: datos sísmicos del IGN

El IGN también publica datos sísmicos en formato XML a través de su servicio FDSN. Vamos a leer ese XML y a extraer información de los terremotos más recientes.

La URL siguiente devuelve los 50 terremotos más recientes de la Península Ibérica en XML:
```
https://www.ign.es/ign/RssTools/sismologia.xml
```

**Parte A:** descarga el XML y explora su estructura.


````python
import requests
import xml.etree.ElementTree as ET

url = 'https://www.ign.es/ign/RssTools/sismologia.xml'
respuesta = requests.get(url, timeout=15)
xml_texto = respuesta.text

# Guardamos el XML por si queremos inspeccionarlo
with open('sismos_ign.xml', 'w', encoding='utf-8') as f:
    f.write(xml_texto)

# Parseamos
root = ET.fromstring(xml_texto)
print('Etiqueta raíz:', root.tag)
print('Namespaces detectados en el XML (primeros 500 chars):')
print(xml_texto[:500])
````

**Parte B:** extrae los datos relevantes y conviértelos en un DataFrame.


````python
import re
from datetime import datetime
import pandas as pd

# El feed es RSS GeoRSS: fecha y magnitud aparecen en el texto, y las
# coordenadas están en etiquetas geo:lat y geo:long.
ns = {'geo': 'http://www.w3.org/2003/01/geo/wgs84_pos#'}
registros = []

for item in root.findall('.//item'):
    titulo = item.findtext('title', default='')
    descripcion = item.findtext('description', default='')
    texto = f'{titulo} {descripcion}'

    fecha_match = re.search(r'\d{1,2}/\d{1,2}/\d{4}\s+\d{1,2}:\d{2}:\d{2}', texto)
    magnitud_match = re.search(r'magnitud\s+([\d.,]+)', descripcion, re.IGNORECASE)
    latitud = item.findtext('geo:lat', namespaces=ns)
    longitud = item.findtext('geo:long', namespaces=ns)

    if not (fecha_match and magnitud_match and latitud and longitud):
        continue

    registros.append({
        'fecha': datetime.strptime(fecha_match.group(), '%d/%m/%Y %H:%M:%S'),
        'latitud': float(latitud),
        'longitud': float(longitud),
        'profundidad_km': None,  # El RSS no publica la profundidad.
        'magnitud': float(magnitud_match.group(1).replace(',', '.')),
    })

df_sismos = pd.DataFrame(registros)
print(f'Terremotos extraídos: {len(df_sismos)}')
print(df_sismos.head())
print('La profundidad no aparece en este feed RSS, se necesitaría otra fuente para obtenerla.')
````

````python
import requests
import pandas as pd
from pathlib import Path

parquet_path = 'pokedex_500.parquet'

if not Path(parquet_path).exists():
    print('Descargando 500 Pokemon...')
    registros = []
    for i in range(1, 501):
        datos = requests.get(f'https://pokeapi.co/api/v2/pokemon/{i}').json()
        tipos = [t['type']['name'] for t in datos['types']]
        stats = {s['stat']['name']: s['base_stat'] for s in datos['stats']}
        registros.append({
            'id':         datos['id'],
            'nombre':     datos['name'],
            'peso':       datos['weight'],
            'altura':     datos['height'],
            'tipo_1':     tipos[0],
            'tipo_2':     tipos[1] if len(tipos) > 1 else None,
            'hp':         stats.get('hp', 0),
            'ataque':     stats.get('attack', 0),
            'defensa':    stats.get('defense', 0),
            'sp_ataque':  stats.get('special-attack', 0),
            'sp_defensa': stats.get('special-defense', 0),
            'velocidad':  stats.get('speed', 0),
        })
    pd.DataFrame(registros).to_parquet(parquet_path, index=False)
    print('Guardado en', parquet_path)
else:
    print('Fichero ya existe, cargando...')

df_pandas = pd.read_parquet(parquet_path)
print(f'Dataset: {len(df_pandas)} filas, {len(df_pandas.columns)} columnas')
````

**Parte B:** realiza las mismas operaciones con pandas y con Dask, y compara.


````python
import dask.dataframe as dd
import time

df_dask = dd.from_pandas(df_pandas, npartitions=4)

print('Particiones de Dask:', df_dask.npartitions)
print('Tipo de df_dask:', type(df_dask))
print()

# Operacion 1: media de estadisticas por tipo
t0 = time.perf_counter()
res_pandas = df_pandas.groupby('tipo_1')[['hp','ataque','defensa','velocidad']].mean().round(1)
t1 = time.perf_counter()
print(f'Pandas  — media por tipo: {(t1-t0)*1000:.2f} ms')
print(res_pandas.sort_values('ataque', ascending=False).head(5))

t0 = time.perf_counter()
res_dask = df_dask.groupby('tipo_1')[['hp','ataque','defensa','velocidad']].mean().compute().round(1)
t1 = time.perf_counter()
print(f'Dask    — media por tipo: {(t1-t0)*1000:.2f} ms')
````

## Ejercicio 3.3 — MapReduce a mano

MapReduce es el paradigma detrás de Hadoop y de la idea central de Spark. Aunque en la práctica usarías frameworks como Spark, entender la lógica map → shuffle → reduce es fundamental.

Vamos a implementar dos problemas clásicos con Python puro (sin ningún framework distribuido).

**Problema A — Contar tipos de Pokémon:**  
Usando el dataset de 500 Pokémon, calcula cuántos Pokémon hay de cada tipo con la lógica MapReduce: primero emite pares `(tipo, 1)` para cada Pokémon, luego agrupa por tipo y suma.


````python
from collections import defaultdict

# ── FASE MAP ─────────────────────────────────────────────────────────────
# Emitir (tipo_1, 1) para cada Pokemon
def map_tipo(pokemon):
    return (pokemon['tipo_1'], 1)

pares_mapeados = [map_tipo(row) for row in df_pandas.to_dict('records')]
print('Primeros 5 pares mapeados:', pares_mapeados[:5])

# ── FASE SHUFFLE (agrupar por clave) ──────────────────────────────────────
grupos = defaultdict(list)
for clave, valor in pares_mapeados:
    grupos[clave].append(valor)

# ── FASE REDUCE ──────────────────────────────────────────────────────────
def reduce_suma(clave, valores):
    return (clave, sum(valores))

resultado = dict(reduce_suma(k, v) for k, v in grupos.items())
resultado_ordenado = sorted(resultado.items(), key=lambda x: x[1], reverse=True)

print('\nConteo de Pokemon por tipo (MapReduce):')
for tipo, cantidad in resultado_ordenado:
    print(f'  {tipo:<12}: {cantidad}')
````

**Problema B — Tu turno:**  
Usando el mismo dataset, calcula con MapReduce la **suma total de puntos de ataque** por tipo de Pokémon. La fase map debe emitir `(tipo_1, ataque)` y la fase reduce debe sumar los valores de cada grupo.


````python
from collections import defaultdict

# Fase MAP: emitir (tipo, ataque) para cada Pokemon.
def map_ataque(pokemon):
    return (pokemon['tipo_1'], pokemon['ataque'])

pares_mapeados = [map_ataque(row) for row in df_pandas.to_dict('records')]
print('Primeros 5 pares mapeados:', pares_mapeados[:5])

# Fase SHUFFLE: agrupar los ataques por tipo.
grupos = defaultdict(list)
for tipo, ataque in pares_mapeados:
    grupos[tipo].append(ataque)

# Fase REDUCE: sumar los ataques de cada tipo.
def reduce_suma(tipo, ataques):
    return (tipo, sum(ataques))

resultado = [reduce_suma(tipo, ataques) for tipo, ataques in grupos.items()]
resultado_ordenado = sorted(resultado, key=lambda par: par[1], reverse=True)

# Mostrar los 5 tipos con mayor ataque total.
print('\nTipos con mayor ataque total:')
for tipo, ataque_total in resultado_ordenado[:5]:
    print(f'  {tipo:<12}: {ataque_total}')
````

## Ejercicio 3.4 — Simula un stream sísmico

Los sistemas de alerta temprana de terremotos procesan datos de sensores en tiempo real y deben detectar eventos relevantes en milisegundos. Vamos a simular ese flujo con Python.

El código de abajo genera un stream de lecturas sísmicas. Tu tarea es procesar ese stream con una **ventana deslizante** de los últimos 10 registros y emitir una alerta cuando la media de magnitud supere 2.5.


````python
import random
from collections import deque

random.seed(42)

VENTANA = 10
UMBRAL = 2.5
N_TOTAL = 100

def generar_lectura_sismica(sismo_activo=False):
    """Simula una lectura normal o una lectura durante un sismo."""
    if sismo_activo or random.random() < 0.05:
        magnitud = round(random.uniform(2.5, 5.5), 1)
    else:
        magnitud = round(random.uniform(0.1, 1.8), 1)
    return {
        'magnitud': magnitud,
        'latitud': round(random.uniform(36.8, 37.5), 4),
        'longitud': round(random.uniform(-3.8, -3.2), 4),
        'profundidad': random.randint(5, 30),
    }

ventana = deque(maxlen=VENTANA)
duracion_sismo = 0
alertas = 0
alerta_activa = False

print(f'Procesando {N_TOTAL} lecturas sismicas...')
print('-' * 55)

for i in range(N_TOTAL):
    # Un episodio dura una ventana para que varias lecturas consecutivas reflejen el sismo.
    if duracion_sismo == 0 and random.random() < 0.05:
        duracion_sismo = VENTANA

    sismo_activo = duracion_sismo > 0
    if sismo_activo:
        duracion_sismo -= 1

    lectura = generar_lectura_sismica(sismo_activo)
    ventana.append(lectura['magnitud'])
    media_ventana = sum(ventana) / len(ventana)

    if len(ventana) == VENTANA and media_ventana >= UMBRAL and not alerta_activa:
        print(f'ALERTA: media={media_ventana:.2f} - '
              f'magnitud={lectura["magnitud"]} '
              f'lat={lectura["latitud"]} lon={lectura["longitud"]}')
        alertas += 1
        alerta_activa = True
    elif media_ventana < UMBRAL:
        alerta_activa = False

    if lectura['magnitud'] >= 2.5:
        print(f'  Lectura {i+1:>3}: magnitud={lectura["magnitud"]} '
              f'lat={lectura["latitud"]} lon={lectura["longitud"]}')

print('-' * 55)
print(f'Total alertas emitidas: {alertas}')
````
