### 2.2 Distribución de las variables numéricas

Visualiza la distribución de cada variable numérica. Identifica si son simétricas, sesgadas o bimodales.

````python
# TU CÓDIGO AQUÍ: histograma + boxplot para cada variable numérica
# Puedes usar la función diagnostico_visual() de los cuadernos de práctica
# o construir tus propios gráficos.

cols_num = df.select_dtypes(include='number').columns.tolist()
fig, axes = plt.subplots(len(cols_num), 2, figsize=(12, 4*len(cols_num)))
for i, col in enumerate(cols_num):
    axes[i,0].hist(df[col].dropna(), bins=30, edgecolor='white')
    axes[i,0].set_title(f'Histograma de {col}')
    axes[i,0].set_xlabel(col)
    axes[i,0].set_ylabel('Frecuencia')
    axes[i,1].boxplot(df[col].dropna(), vert=False)
    axes[i,1].set_title(f'Boxplot de {col}')
    axes[i,1].set_xlabel(col)
plt.tight_layout()
plt.show()
````

### 2.3 Relaciones entre variables

Analiza las relaciones más relevantes: entre variables numéricas entre sí, y entre variables numéricas y categóricas.

````python
# ── Mapa de correlaciones (variables numéricas) ───────────────────────────
# TU CÓDIGO AQUÍ
corr = df[cols_num].corr()
fig, ax = plt.subplots(figsize=(12, 9))
sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0,
            linewidths=0.5, ax=ax)
ax.set_title('Mapa de correlaciones')
plt.tight_layout()
plt.show()

# ── Relación entre una variable numérica y una categórica ─────────────────
# TU CÓDIGO AQUÍ: elige el par más relevante para tu dataset
sns.boxplot(data=df, x='RainTomorrow', y='Humidity3pm')
plt.title('Humedad a las 3pm según si llueve al día siguiente')
plt.xlabel('RainTomorrow')
plt.ylabel('Humidity3pm')
plt.show()
````

````python
# ── Gráfico de dispersión del par más correlacionado ─────────────────────
# TU CÓDIGO AQUÍ: identifica el par con |r| más alto y traza un scatter plot
fig, ax = plt.subplots(figsize=(7, 5))
ax.scatter(df['MaxTemp'], df['Temp3pm'], alpha=0.5, s=20)
ax.set_xlabel('MaxTemp')
ax.set_ylabel('Temp3pm')
ax.set_title('MaxTemp vs Temp3pm')
plt.show()
````
