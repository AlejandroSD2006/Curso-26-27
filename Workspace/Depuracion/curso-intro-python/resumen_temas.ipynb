# Guía de estudio: Curso de Programación en Python

Este resumen recorre los nueve temas del curso. La idea es explicar los conceptos fundamentales, distinguir términos que suelen confundirse y mostrar cómo cada unidad prepara la siguiente. No sustituye los ejemplos y ejercicios de los notebooks originales: sirve como mapa para repasar y entender el vocabulario.

## Tema 1 - Introducción y generalidades

Python es un lenguaje de programación de alto nivel, interpretado, dinámico y portable. En una instalación habitual, el intérprete traduce el código fuente a *bytecode* y la máquina virtual de Python ejecuta ese bytecode. El programador trabaja con el código Python y no necesita compilar manualmente un ejecutable para cada sistema. El tipado dinámico significa que los objetos tienen tipo, pero no hace falta declarar de antemano un tipo fijo para cada variable.

El desarrollo suele ser incremental: se escribe una parte pequeña, se ejecuta y se observa el resultado. El intérprete interactivo permite probar instrucciones y consultar objetos; un *script* es un archivo `.py` que ejecuta una secuencia de instrucciones; un módulo es código organizado para poder reutilizarlo desde otros programas. Jupyter combina celdas de texto, código y resultados en un documento interactivo, por lo que es útil para aprendizaje, experimentación y documentación.

Los paquetes externos se instalan normalmente con `pip`. Un entorno virtual (`venv` o un entorno de Conda) mantiene separadas las dependencias de un proyecto, para reducir conflictos entre versiones. En Python, la indentación delimita bloques de código y forma parte de la sintaxis; por convención se usan cuatro espacios. El curso utiliza Python 3.

## Tema 2 - Sentencias, objetos, tipos y variables

Un **objeto** es una entidad que existe durante la ejecución: puede ser un número, una lista, una función o una instancia de una clase. Cada objeto tiene un tipo, un valor y una identidad; además puede ofrecer **atributos**, que son datos o métodos asociados al objeto. Se accede a un atributo con la notación `objeto.atributo`; `dir` y `help` ayudan a inspeccionar objetos y consultar su documentación. Una **variable** no es una caja que contenga necesariamente una copia: es un nombre ligado a un objeto. Por eso, el tipo corresponde al objeto y un mismo nombre puede ligarse después a objetos de tipos distintos.

Entre los tipos incorporados están `int`, `float` y `complex` para números; `bool` para `True` y `False`; `None` para indicar ausencia de un valor útil; `str` para texto; listas, tuplas y conjuntos para colecciones; y `dict` para asociaciones entre claves y valores. Una **expresión** produce un valor, mientras que una **sentencia** realiza una acción o controla el programa. La asignación liga un nombre a un objeto, y `if` ejecuta bloques distintos según una condición.

La **mutabilidad** indica si un objeto puede cambiar después de crearse: una lista es mutable, mientras que una tupla o un string son inmutables. Si `b = a` y `a` es una lista, ambos nombres se refieren inicialmente al mismo objeto; modificarlo a través de uno se ve a través del otro. `==` compara valores y `is` comprueba si se trata del mismo objeto. Crear otra lista con los mismos elementos es una copia superficial, no una simple asignación. Python administra la memoria automáticamente y puede liberar objetos que ya no tienen referencias activas.

## Tema 3 - Iterables y bucles (I): secuencias

Un **iterable** es un objeto cuyos elementos pueden recorrerse uno a uno. El bucle `for` obtiene sus elementos sucesivamente y ejecuta el bloque una vez por elemento; no exige conocer de antemano cuántos elementos habrá. Una **secuencia** es un tipo de iterable ordenado que permite acceder a sus posiciones mediante índices enteros y, normalmente, extraer partes mediante *slicing* (`inicio:fin:paso`). Los índices empiezan en cero.

Las **listas** son secuencias mutables: se pueden modificar, añadir o quitar elementos. Las **tuplas** también son secuencias ordenadas, pero inmutables; son útiles cuando el grupo de valores debe permanecer fijo. `range` representa una secuencia inmutable de enteros y se evalúa de forma perezosa: describe el intervalo sin construir de antemano una lista completa, lo que lo hace práctico en bucles.

Los **sets** almacenan elementos únicos, sin duplicados ni un orden posicional en el que debamos confiar; resultan útiles para pertenencia y operaciones de conjuntos como unión e intersección. Los **strings** son secuencias inmutables de caracteres Unicode: se pueden indexar y recorrer, pero una operación como `upper()` devuelve otro string en vez de modificar el original. El operador `in` permite preguntar si un elemento o fragmento pertenece a una colección.

## Tema 4 - Iterables y bucles (II): diccionarios

Un **diccionario** (`dict`) es un mapa que relaciona cada **clave** con un **valor**. Las claves deben ser *hashables* y permanecer estables mientras se usan como claves; en los ejemplos básicos suelen ser números, strings o tuplas inmutables. Los valores pueden ser objetos de cualquier tipo. Los diccionarios permiten recuperar información por una clave descriptiva, sin buscar manualmente una posición numérica. Se pueden crear, consultar, añadir, actualizar y borrar asociaciones; la expresión `clave in diccionario` comprueba si esa clave existe.

Al recorrer directamente un diccionario se recorren sus claves. Los métodos `keys()`, `values()` e `items()` devuelven **vistas** dinámicas: reflejan los cambios posteriores del diccionario; `items()` permite recorrer pares clave-valor. `enumerate(iterable)` produce pares índice-elemento y `zip(a, b, ...)` combina elementos correspondientes de varios iterables, deteniéndose cuando se agota el más corto. Ambos se evalúan de forma perezosa. `while` repite un bloque mientras una condición sea verdadera y es apropiado cuando el final depende de una condición, no solo de recorrer una colección.

Un ejemplo de modelado es representar un grafo con un diccionario: cada nodo es una clave y su valor es la lista de nodos vecinos o conexiones. Elegir una estructura de datos adecuada hace que operaciones como encontrar las conexiones de un nodo resulten directas.

## Tema 5 - Entrada y salida

Los strings permiten trabajar con texto. Las secuencias de escape, como `\n` o `\t`, representan saltos de línea y tabuladores; un *raw string* conserva normalmente las barras invertidas sin interpretarlas como escapes. `str(obj)` produce una representación legible del objeto y `repr(obj)` una representación orientada a mostrarlo de forma inequívoca o útil para depuración. Ninguna de las dos es, por sí sola, un formato general de almacenamiento. Las *f-strings* insertan expresiones y valores formateados directamente en una cadena.

`open(ruta, modo)` abre un archivo y devuelve un objeto de archivo. Los modos más habituales son `r` para lectura, `w` para escribir desde cero y `a` para añadir al final; se añade `b` cuando se trabaja con bytes en vez de texto. `read`, `readline` y `readlines` leen el contenido completo, una línea o todas las líneas en una lista; `write` escribe texto. Usar `with open(...) as fichero` cierra el archivo automáticamente incluso si ocurre una excepción. Para archivos grandes, recorrer el objeto archivo línea a línea evita cargar todo de una vez.

La **serialización** transforma datos en una representación que puede guardarse y reconstruirse. JSON es texto legible y ampliamente interoperable, aunque solo representa directamente ciertos tipos de datos. `pickle` puede guardar más objetos propios de Python, pero es específico de Python y cargar un pickle no confiable puede ejecutar código malicioso: no debe usarse con datos de origen desconocido. `eval` evalúa una expresión Python escrita como texto y `exec` ejecuta instrucciones; ambos son peligrosos con contenido no confiable. `input()` siempre devuelve un string, por lo que hay que convertirlo, por ejemplo con `int()`. `sys.stdin` permite leer como un archivo, especialmente cuando la entrada llega mediante una tubería o redirección.

## Tema 6 - Funciones y módulos

Una **función** es un bloque reutilizable definido con `def` que se ejecuta cuando se llama. Sus parámetros reciben argumentos y `return` entrega un resultado; si la función termina sin `return`, devuelve `None`. Los argumentos pueden pasarse por posición o por nombre, tener valores predeterminados, agruparse con `*args` en una tupla o recogerse con `**kwargs` en un diccionario. Python pasa argumentos por asignación: el parámetro local queda ligado al mismo objeto recibido, así que una mutación de un objeto mutable puede observarse fuera de la función, aunque reasignar el parámetro no reasigna el nombre del llamador.

El **polimorfismo** permite que una misma operación funcione con objetos diferentes. En Python es común el *duck typing*: importa que un objeto ofrezca las operaciones necesarias, no que pertenezca a una clase concreta. Las anotaciones (`x: int`, `-> str`) documentan los tipos esperados y ayudan a las herramientas, pero Python no las valida automáticamente al llamar a la función. Las funciones son objetos: se pueden guardar en variables, pasar como argumentos y devolver desde otras funciones. La recursividad ocurre cuando una función se llama a sí misma y necesita un caso base que detenga las llamadas.

Un **namespace** es un espacio que relaciona nombres con objetos; el **scope** o alcance determina desde qué partes del programa se pueden usar esos nombres. Las funciones tienen variables locales y los módulos crean su propio espacio de nombres. Un **módulo** suele ser un archivo `.py` reutilizable; un **paquete** agrupa módulos en un directorio, normalmente identificado por `__init__.py`. `import` hace accesible el contenido del módulo. La condición `if __name__ == '__main__':` permite que un archivo tenga código de uso como script que no se ejecuta cuando se importa como módulo. Un **docstring** es una cadena inicial de un módulo, clase o función que documenta su propósito y aparece, por ejemplo, con `help()`.

## Tema 7 - Clases y objetos

La **programación orientada a objetos** organiza datos y operaciones relacionadas en objetos. Una **clase** define una estructura y comportamiento común; una **instancia** es un objeto concreto creado a partir de esa clase. Los atributos guardan estado y los métodos implementan operaciones. `__init__` inicializa la instancia y `self` es el parámetro que referencia a la instancia sobre la que se está trabajando; Python lo pasa automáticamente al invocar un método a través del objeto.

La **encapsulación** consiste en ofrecer una interfaz de uso y mantener los detalles internos como una decisión de implementación. Python no impone privacidad estricta como algunos lenguajes: un nombre que empieza con `_` es una convención para indicar que es interno. La **herencia** permite definir una clase especializada a partir de otra, reutilizando o ampliando su comportamiento. La sobrescritura de métodos permite que una subclase implemente una operación de otra manera; el **polimorfismo** permite usar objetos distintos a través de una interfaz común. `classmethod` recibe la clase como primer argumento y `staticmethod` es una función agrupada en la clase sin recibir automáticamente ni instancia ni clase.

Las pruebas verifican que el comportamiento observado coincide con el esperado. En el **desarrollo guiado por pruebas** (*test-driven development*), se escriben o utilizan pruebas para definir los casos esperados y se desarrolla la implementación hasta que los supera. Esto ayuda a detectar regresiones y concreta qué significa que una función o clase funcione correctamente.

## Tema 8 - Otros recursos avanzados

Una **excepción** es un objeto que señala una situación que altera el flujo normal del programa. `try` contiene el código que puede fallar; `except` captura tipos de error concretos; `else` se ejecuta si no hubo excepción y `finally` se ejecuta al terminar en ambos casos, por ejemplo para limpiar recursos. `raise` lanza una excepción explícitamente. Se pueden crear excepciones propias heredando de `Exception`, para representar errores específicos del dominio. Conviene capturar errores concretos en vez de ocultar indiscriminadamente cualquier fallo.


## Tema 8 - Otros recursos avanzados (continuación)

Una **comprehension** construye de forma concisa una colección recorriendo un iterable y transformando o filtrando sus elementos. Hay comprensiones de listas, diccionarios y conjuntos. Una expresión generadora tiene una forma parecida, pero produce los valores solo cuando se solicitan, en lugar de construir toda la colección inmediatamente.

Un **decorador** es una función que recibe otra función y devuelve una versión modificada o envuelta. Se usa, por ejemplo, para añadir medición de tiempo o comprobaciones sin reescribir la función original. La sintaxis `@decorador` aplica esa transformación en el momento de definirla.

Un **iterable** permite recorrer elementos; un **iterador** es el objeto que avanza y entrega el siguiente elemento. `iter(objeto)` obtiene un iterador y `next(iterador)` pide el siguiente valor. Cuando ya no quedan elementos, se señala el final con `StopIteration`. Un **generador** crea iteradores perezosos de manera sencilla: una función con `yield` produce un valor y queda suspendida hasta que se solicita el siguiente. La evaluación perezosa puede ahorrar memoria cuando no hace falta materializar una secuencia completa.

Un **context manager** gestiona la adquisición y liberación de un recurso. La instrucción `with` asegura que la limpieza se produzca al abandonar el bloque, incluso si ocurre un error; cerrar un archivo es el ejemplo habitual. Así se evita depender de que el programador recuerde liberar manualmente cada recurso.

## Tema 9 - El ecosistema Python

La **librería estándar** se distribuye con Python y ofrece herramientas generales sin instalar paquetes externos. `sys` da acceso a información y entradas/salidas del intérprete; `os`, `pathlib`, `glob`, `shutil` y `tempfile` ayudan con el sistema de archivos; `subprocess` permite lanzar comandos externos. `threading` y `multiprocessing` ofrecen modelos distintos de concurrencia, mientras que `asyncio` organiza operaciones de entrada/salida asíncronas. `socket` proporciona operaciones de red de bajo nivel.

Para cálculos y programación general, `math`, `random` y `statistics` cubren matemáticas, aleatoriedad y estadística básica; `itertools`, `functools` y `operator` ayudan a trabajar con iteradores y funciones. `pdb` sirve para depurar paso a paso, `cProfile` para perfilar, `timeit` para medir tiempos y `unittest` para crear pruebas. `pytest` es un paquete externo muy popular que facilita la ejecución y escritura de pruebas. Conviene elegir cada módulo según la tarea, y no incorporar una dependencia externa si la librería estándar ya resuelve bien la necesidad.

El ecosistema científico amplía estas capacidades. **NumPy** ofrece arrays y operaciones numéricas eficientes; **SciPy** añade algoritmos científicos como optimización, integración y álgebra lineal. **Pandas** organiza y transforma datos tabulares. **Matplotlib** es una base general para gráficos; **Seaborn** facilita gráficos estadísticos y **Bokeh** y **Plotly** permiten visualizaciones interactivas. **SymPy** trabaja con matemáticas simbólicas y **statsmodels** con modelos y análisis estadísticos.

En aprendizaje automático, **scikit-learn** ofrece algoritmos generales para preparar datos, entrenar modelos y evaluarlos; **TensorFlow** y **PyTorch** se usan para redes neuronales y cálculo acelerado, mientras que **Keras** proporciona una interfaz de alto nivel. Para procesamiento del lenguaje natural se utilizan, entre otras, **NLTK**, **spaCy** y **Gensim**; en biología, **BioPython** y **scikit-bio**; en astronomía y física, **Astropy**, **PyROOT** y herramientas de **Scikit-HEP**. Para trabajar con archivos HDF5 se encuentran **h5py** y **PyTables**. Son familias de herramientas, no una lista de requisitos: la elección depende del problema, el formato de los datos y las necesidades de rendimiento o visualización.

## Cómo se conectan los temas

El curso avanza desde cómo se ejecutan los programas y qué son los objetos (temas 1 y 2) hasta cómo representar y recorrer colecciones (temas 3 y 4). La entrada y salida permite trabajar con información guardada o proporcionada por el usuario (tema 5); las funciones y los módulos ayudan a organizar y reutilizar ese trabajo (tema 6). Las clases amplían esa organización cuando se modelan entidades con estado y comportamiento (tema 7). Las excepciones, los generadores y los gestores de contexto aportan control, eficiencia y limpieza de recursos (tema 8). Finalmente, la librería estándar y los paquetes externos permiten aplicar todos esos fundamentos a problemas especializados (tema 9).

## Glosario breve

- **Objeto:** entidad en ejecución con tipo, valor, identidad y atributos.
- **Iterable:** objeto que permite solicitar sus elementos sucesivamente.
- **Iterador:** objeto que mantiene el avance y entrega el siguiente elemento.
- **Mutable / inmutable:** que puede / no puede cambiar su contenido después de crearse.
- **Referencia:** vínculo de un nombre con un objeto; no implica crear una copia.
- **Namespace / scope:** espacio de nombres / reglas que determinan dónde son visibles los nombres.
- **Serialización:** conversión de datos a una representación que se puede almacenar y reconstruir.
- **Excepción:** objeto que comunica un error o una condición especial y altera el flujo normal.
- **Evaluación perezosa:** producir o calcular cada valor solo cuando se necesita.
- **Paquete:** agrupación organizada de módulos reutilizables.
- **Polimorfismo:** posibilidad de usar objetos distintos mediante operaciones o interfaces comunes.
