En el mundo real los CSV raramente están limpios. A continuación tienes tres cadenas de texto que simulan ficheros CSV con distintos problemas. Tu tarea es diagnosticar cada uno e identificar qué parámetros de `read_csv` necesitas para leerlos correctamente.

**Parte A — Diagnóstico:** ejecuta las celdas e identifica qué está mal en cada CSV (separador incorrecto, encoding, columna mal inferida, valores nulos no reconocidos…).


````python
import pandas as pd
import io

# CSV 1: aparentemente normal
csv1 = """\
id;nombre;precio;stock
1;Ratón inalámbrico;29,99;15
2;Teclado mecánico;89,50;8
3;Monitor 27";349,00;3
4;Auriculares BT;59,95;N/A
5;Webcam HD;45,00;0
"""

# Intento de lectura 'ingenuo'
df1_mal = pd.read_csv(io.StringIO(csv1), sep=';', decimal=',')
df1_mal.info()
print('CSV 1 — lectura sin parámetros:')
print(df1_mal)
print(df1_mal.dtypes)
````

**Preguntas sobre el CSV 1:**

1. ¿Qué está mal con la columna `precio`? ¿Por qué no es de tipo `float64`?
2. ¿Qué indica el valor `N/A` en la columna `stock`?
3. ¿Qué parámetro de `read_csv` usarías para indicar el separador correcto?

*Responde aquí:*


*Escribe aquí tus respuestas:*

1. Si se lee sin indicar nada, pandas no separa las columnas (espera comas y aquí el separador es punto y coma) y el precio, que usa la coma como decimal, se leería como texto y no como float64. Con sep=';' y decimal=',' ya sale como float64.

2. Que a ese producto le falta el dato del stock, es un valor nulo. Pandas ya lo reconoce como nulo, pero al haber un nulo la columna pasa a float64 (15.0, 8.0...) en vez de entero, por eso se usa Int64.

3. El parámetro sep, en este caso sep=';'.

````python
# Leemos el CSV 1 con el separador correcto, la coma como decimal y stock como entero con nulos
df1_bien = pd.read_csv(
    io.StringIO(csv1),
    sep=';',
    decimal=',',
    na_values=['N/A'],
    dtype={'stock': 'Int64'}
)
print('CSV 1 — lectura correcta:')
print(df1_bien)
print(df1_bien.dtypes)
````

**Parte C — Lectura eficiente con chunksize:**  
Imagina que el fichero fuera enorme. Lee los mismos datos en chunks de 200 filas y calcula el total de sismos con profundidad > 100 km sin cargar todo en memoria a la vez.


````python
# Leemos por trozos de 200 filas y vamos sumando los sismos de más de 100 km de profundidad
def contar_profundos(fuente):
    total = 0
    for chunk in pd.read_csv(fuente, chunksize=200, parse_dates=['time']):
        total += (chunk['depth'] > 100).sum()
    return total

try:
    total_profundos = contar_profundos(url)
except Exception:
    # Sin conexión usamos la copia local
    total_profundos = contar_profundos('../Bloque_2_CSV_y_Excel/datos/sismos_usgs_2024.csv')
print(f'Sismos con profundidad > 100 km: {total_profundos}')
````

## Ejercicio 2.3 — De CSV a Excel con múltiples hojas

Retomamos el dataset de sismos del ejercicio anterior. Un geólogo te pide un fichero Excel con:

- **Hoja 1** (`Todos`): todos los sismos con las columnas `time`, `latitude`, `longitude`, `depth`, `mag`, `place`.
- **Hoja 2** (`Fuertes`): solo los de magnitud ≥ 6.0, ordenados de mayor a menor magnitud.
- **Hoja 3** (`Resumen`): tabla con el número de sismos y magnitud media agrupados por `magType`.

El Excel debe tener la primera fila congelada y la columna de magnitud resaltada en amarillo en la hoja `Fuertes`.


````python
import pandas as pd
from openpyxl.styles import PatternFill

COLS = ['time', 'latitude', 'longitude', 'depth', 'mag', 'place']
excel_path = 'sismos_informe.xlsx'

# Hoja 1: todos los sismos (quitamos la zona horaria porque Excel no la admite)
df_todos = df_sismos[COLS].copy()
df_todos['time'] = df_todos['time'].dt.tz_localize(None)

# Hoja 2: solo los fuertes, de mayor a menor magnitud
df_fuertes = df_todos[df_todos['mag'] >= 6.0].sort_values('mag', ascending=False)

# Hoja 3: número de sismos y magnitud media por tipo de magnitud
df_resumen = (df_sismos.groupby('magType')['mag']
              .agg(n_sismos='count', mag_media='mean')
              .round(2)
              .reset_index())

# Escribimos las tres hojas
with pd.ExcelWriter(excel_path, engine='openpyxl') as writer:
    df_todos.to_excel(writer, sheet_name='Todos', index=False)
    df_fuertes.to_excel(writer, sheet_name='Fuertes', index=False)
    df_resumen.to_excel(writer, sheet_name='Resumen', index=False)

    # Congelamos la primera fila de cada hoja
    for ws in writer.book.worksheets:
        ws.freeze_panes = 'A2'

    # Columna de magnitud en amarillo en la hoja Fuertes
    amarillo = PatternFill(fill_type='solid', fgColor='FFFF00')
    ws = writer.sheets['Fuertes']
    col_mag = list(df_fuertes.columns).index('mag') + 1
    for fila in range(1, ws.max_row + 1):
        ws.cell(row=fila, column=col_mag).fill = amarillo

# Verificación: leemos el Excel de vuelta
import openpyxl
wb = openpyxl.load_workbook(excel_path)
print(f'Hojas: {wb.sheetnames}')
for nombre in wb.sheetnames:
    print(f'  {nombre}: {wb[nombre].max_row - 1} filas, primera fila congelada: {wb[nombre].freeze_panes}')
print(df_resumen)
````

## Ejercicio 3.1 — Benchmark CSV vs formatos binarios

Vamos a generar un dataset realista y comparar empíricamente cuánto espacio y tiempo tarda cada formato. Así verás por qué Parquet domina los pipelines de ML.

**Parte A — Genera y guarda:**

## Ejercicio 3.2 — Projection y filter pushdown en Parquet

Una de las ventajas más importantes de Parquet es que permite leer solo las columnas que necesitas (*projection pushdown*) y solo las filas que cumplen un filtro (*filter pushdown*), sin tocar el resto del fichero.

**Parte A:** mide el impacto del projection pushdown.


> **Antes de empezar la Parte B**: el filtro solo puede saltar bloques si el fichero tiene varios *row groups* y los datos están ordenados por la columna del filtro (bloque 3, apartado 3.3). `benchmark_zstd.parquet` tiene un único row group, así que guarda primero una versión preparada:
>
> ```python
> df.sort_values('categoria').to_parquet('benchmark_ordenado.parquet',
>                                        compression='zstd', row_group_size=20_000, index=False)
> ```

````python
import pandas as pd, time, os

# Lectura completa (todas las columnas)
t0 = time.perf_counter()
df_full = pd.read_parquet('benchmark_zstd.parquet')
t_full = time.perf_counter() - t0
mem_full = df_full.memory_usage(deep=True).sum()

# Lectura solo de 3 columnas
t0 = time.perf_counter()
df_proj = pd.read_parquet('benchmark_zstd.parquet', columns=['id_pedido', 'categoria', 'total'])
t_proj = time.perf_counter() - t0
mem_proj = df_proj.memory_usage(deep=True).sum()

print(f'Lectura completa (10 cols): {t_full*1000:.0f} ms  {mem_full/1e6:.1f} MB RAM')
print(f'Projection (3 cols):        {t_proj*1000:.0f} ms  {mem_proj/1e6:.1f} MB RAM')
````

**Parte B:** filter pushdown con pyarrow.


````python
import pyarrow.parquet as pq

# Primero guardamos una versión ordenada por categoría y con varios row groups
df.sort_values('categoria').to_parquet('benchmark_ordenado.parquet',
                                       compression='zstd', row_group_size=20_000, index=False)

# Con filter pushdown (pyarrow lee solo los row groups necesarios)
t0 = time.perf_counter()
df_filt_pq = pq.read_table('benchmark_ordenado.parquet',
                           filters=[('categoria', '=', 'Electronica')]).to_pandas()
t_pq = time.perf_counter() - t0

# Sin filter pushdown (carga todo y filtra en pandas)
t0 = time.perf_counter()
df_filt_pd = pd.read_parquet('benchmark_ordenado.parquet')
df_filt_pd = df_filt_pd[df_filt_pd['categoria'] == 'Electronica']
t_pd = time.perf_counter() - t0

print(f'Filter pushdown (pyarrow): {t_pq*1000:.0f} ms  {len(df_filt_pq)} filas')
print(f'Cargar todo + filtrar:     {t_pd*1000:.0f} ms  {len(df_filt_pd)} filas')

# Row groups totales y cuántos tiene que leer el filtro
# (miramos el mínimo y el máximo de categoría que guarda cada row group)
pf = pq.ParquetFile('benchmark_ordenado.parquet')
col = pf.schema_arrow.get_field_index('categoria')
n_rg = pf.metadata.num_row_groups
a_leer = 0
for i in range(n_rg):
    est = pf.metadata.row_group(i).column(col).statistics
    if est.min <= 'Electronica' <= est.max:
        a_leer += 1
print(f'Row groups totales: {n_rg}  |  row groups que lee el filtro: {a_leer}')
````

Un grupo de investigación de la UGR quiere almacenar un dataset de señales de EEG (electroencefalograma) para entrenar un clasificador de estados cognitivos. Cada señal es un array de 64 canales × 256 muestras (float32).

Tu tarea es crear un fichero HDF5 con la siguiente estructura:

```
eeg_dataset.h5
├── train/
│   ├── X   [1000 × 64 × 256, float32, compresión gzip]
│   └── y   [1000, int32]           ← etiquetas 0-3
├── test/
│   ├── X   [200 × 64 × 256, float32]
│   └── y   [200, int32]
└── metadatos/  (grupo con atributos)
    ├── @n_clases = 4
    ├── @etiquetas = 'reposo,atencion,relajacion,esfuerzo'
    └── @frecuencia_muestreo_hz = 256
```


````python
import h5py
import numpy as np

rng = np.random.default_rng(7)

# Datos simulados de EEG
X_train = rng.standard_normal((1000, 64, 256)).astype('float32')
y_train = rng.integers(0, 4, 1000).astype('int32')
X_test  = rng.standard_normal((200, 64, 256)).astype('float32')
y_test  = rng.integers(0, 4, 200).astype('int32')

# Creamos el fichero HDF5 con la estructura indicada
with h5py.File('eeg_dataset.h5', 'w') as hf:
    # Grupo train (X comprimido con gzip)
    grp_train = hf.create_group('train')
    grp_train.create_dataset('X', data=X_train, compression='gzip')
    grp_train.create_dataset('y', data=y_train)

    # Grupo test
    grp_test = hf.create_group('test')
    grp_test.create_dataset('X', data=X_test)
    grp_test.create_dataset('y', data=y_test)

    # Grupo metadatos con atributos
    meta = hf.create_group('metadatos')
    meta.attrs['n_clases'] = 4
    meta.attrs['etiquetas'] = 'reposo,atencion,relajacion,esfuerzo'
    meta.attrs['frecuencia_muestreo_hz'] = 256

print('Fichero eeg_dataset.h5 creado.')
````

**Parte B — Parsea y extrae a DataFrame:**  
El XML del IGN usa el namespace `http://quakeml.org/xmlns/bed/1.2`. Extrae `tiempo`, `latitud`, `longitud`, `profundidad_km` y `magnitud` de cada terremoto y construye un DataFrame.


````python
import pandas as pd
import re
import xml.etree.ElementTree as ET
from datetime import datetime

# Namespaces: el de QuakeML (servicio FDSN) y el de GeoRSS (feed RSS)
NS = {'q': 'http://quakeml.org/xmlns/bed/1.2',
      'geo': 'http://www.w3.org/2003/01/geo/wgs84_pos#'}

import os
if not os.path.exists('sismos_ign.xml'):
    raise FileNotFoundError('Falta sismos_ign.xml: ejecuta antes la Parte A con conexión a internet')

root = ET.parse('sismos_ign.xml').getroot()

registros = []

if root.findall('.//q:event', NS):
    # Formato QuakeML (el que describe el ejercicio)
    for event in root.findall('.//q:event', NS):
        origin = event.find('q:origin', NS)
        mag    = event.find('q:magnitude', NS)
        registros.append({
            'tiempo':     origin.findtext('q:time/q:value', '', NS),
            'latitud':    float(origin.findtext('q:latitude/q:value', 0, NS)),
            'longitud':   float(origin.findtext('q:longitude/q:value', 0, NS)),
            'profundidad_km': float(origin.findtext('q:depth/q:value', 0, NS)) / 1000,
            'magnitud':   float(mag.findtext('q:mag/q:value', 0, NS)) if mag is not None else None,
        })
else:
    # Formato RSS: la fecha y la magnitud van dentro del texto y las coordenadas en etiquetas geo
    # Este feed no trae profundidad, así que esa columna se queda vacía
    for item in root.findall('.//item'):
        descripcion = item.findtext('description', default='')
        fecha = re.search(r'\d{1,2}/\d{1,2}/\d{4}\s+\d{1,2}:\d{2}:\d{2}', descripcion)
        magnitud = re.search(r'magnitud\s+([\d.,]+)', descripcion, re.IGNORECASE)
        if not (fecha and magnitud):
            continue
        registros.append({
            'tiempo':     datetime.strptime(fecha.group(), '%d/%m/%Y %H:%M:%S'),
            'latitud':    float(item.findtext('geo:lat', namespaces=NS)),
            'longitud':   float(item.findtext('geo:long', namespaces=NS)),
            'profundidad_km': None,
            'magnitud':   float(magnitud.group(1)),
        })

df_ign = pd.DataFrame(registros)
print(f'Extraídos: {len(df_ign)} sismos')
print(df_ign.head())
````

````python
import xml.etree.ElementTree as ET

arbol = ET.parse('../Bloque_4_XML/datos/catalogo_libros.xml')
raiz = arbol.getroot()
print('Libros antes:', len(raiz.findall('libro')))

# --- Parte A: modificar ---
for libro in raiz.findall('libro'):
    precio = libro.find('precio')
    stock = int(libro.findtext('stock'))
    # Subimos un 5 % el precio de los libros técnicos
    if libro.findtext('genero') == 'Tecnico':
        precio.text = f'{float(precio.text) * 1.05:.2f}'
    # El atributo disponible tiene que cuadrar con el stock
    libro.set('disponible', 'false' if stock == 0 else 'true')

# --- Parte B: borrar ---
# Primero guardamos los descatalogados y luego los borramos
descatalogados = [l for l in raiz.findall('libro')
                  if int(l.findtext('stock')) == 0 and int(l.findtext('ano')) < 1900]
for libro in descatalogados:
    raiz.remove(libro)
print('Libros borrados:', len(descatalogados))
print('Libros después:', len(raiz.findall('libro')))

# --- Parte C: guardar ---
ET.indent(arbol, space='  ')
arbol.write('catalogo_actualizado.xml', encoding='utf-8', xml_declaration=True)
````

## Ejercicio 6.3 — Seudonimizar antes de compartir

El equipo de ciencia de datos quiere entrenar un modelo de segmentación de clientes con `usuarios_anidado.json`, pero **no debe ver datos personales**. Genera el fichero `usuarios_ml.csv` (y su versión `usuarios_ml.json`) con **una fila por usuario** y estas columnas:

| Columna | Cómo se obtiene |
|---|---|
| `id_seud` | Seudónimo del correo con **HMAC-SHA256** y una clave secreta (primeros 16 caracteres hexadecimales) |
| `rango_edad` | Edad generalizada en tramos: `18-24`, `25-34`, `35-44`, `45-54`, `55+` |
| `provincia_cp` | Solo los **dos primeros dígitos** del código postal |
| `ciudad` | Se mantiene |
| `n_compras`, `gasto_total` | Calculados a partir de `historial_compras` |
| `categorias_favoritas` | Unidas con `;` |

**No** deben aparecer `nombre`, `email`, `calle` ni el `id` original.

La clave secreta **no se escribe en el cuaderno**: se lee de la variable de entorno `ETCDFM_CLAVE_SEUD`. En Windows se define una vez con `setx ETCDFM_CLAVE_SEUD "una-clave-larga"` en la terminal; después hay que cerrar y volver a abrir VS Code para que el cuaderno la vea. Si no existe, el código genera una clave aleatoria para esta sesión.

````python
import hashlib
import hmac
import json
import os
import secrets
import pandas as pd

CLAVE = os.environ.get('ETCDFM_CLAVE_SEUD')
if CLAVE is None:
    CLAVE = secrets.token_hex(32)
    print('Aviso: no hay ETCDFM_CLAVE_SEUD; se usa una clave aleatoria solo para esta sesión.')

def seudonimo(valor: str) -> str:
    """Devuelve un seudónimo estable de 'valor' usando HMAC-SHA256 con CLAVE."""
    return hmac.new(CLAVE.encode(), valor.encode(), hashlib.sha256).hexdigest()[:16]

def rango_edad(edad: int) -> str:
    if edad < 25:
        return '18-24'
    if edad < 35:
        return '25-34'
    if edad < 45:
        return '35-44'
    if edad < 55:
        return '45-54'
    return '55+'

with open('../Bloque_5_JSON/datos/usuarios_anidado.json', encoding='utf-8') as f:
    usuarios = json.load(f)

filas = []
for u in usuarios:
    compras = u.get('historial_compras', [])
    filas.append({
        'id_seud': seudonimo(u['email']),
        'rango_edad': rango_edad(u['edad']),
        'provincia_cp': u['direccion']['codigo_postal'][:2],
        'ciudad': u['direccion']['ciudad'],
        'n_compras': len(compras),
        'gasto_total': round(sum(c['total_eur'] for c in compras), 2),
        'categorias_favoritas': ';'.join(u['preferencias']['categorias_favoritas']),
    })

df_ml = pd.DataFrame(filas)
print(df_ml.head())

# Guardamos en CSV y en JSON
df_ml.to_csv('usuarios_ml.csv', index=False)
df_ml.to_json('usuarios_ml.json', orient='records', force_ascii=False, indent=2)

# Comprobación: en el fichero generado no puede quedar ningún correo ni nombre
texto = df_ml.to_csv(index=False)
fugas = [u['email'] for u in usuarios if u['email'] in texto] + \
        [u['nombre'] for u in usuarios if u['nombre'] in texto]
print('Datos personales encontrados:', fugas if fugas else 'ninguno')
````

````python
def enmascarar_ip(ip: str) -> str:
    # Cambiamos el último número de la IP por 0
    partes = ip.split('.')
    partes[-1] = '0'
    return '.'.join(partes)

with open('../Bloque_5_JSON/datos/eventos.ndjson', encoding='utf-8') as origen, \
     open('eventos_ml.ndjson', 'w', encoding='utf-8') as destino:
    for linea in origen:
        evento = json.loads(linea)
        # Enmascaramos la IP y seudonimizamos el usuario
        evento['ip'] = enmascarar_ip(evento['ip'])
        evento['usuario_id'] = seudonimo(str(evento['usuario_id']))
        destino.write(json.dumps(evento, ensure_ascii=False) + '\n')

print(open('eventos_ml.ndjson', encoding='utf-8').readline())
````

````python
import time
import pandas as pd
import dask.dataframe as dd

RUTA = '../Bloque_3_Formatos_avanzados/datos/ventas_grande.csv'

# --- Parte A ---
ddf = dd.read_csv(RUTA, blocksize='256KB')
print('Particiones:', ddf.npartitions)

# Filtramos (tarjeta y cantidad de 5 o más), agrupamos por región y calculamos todo al final con compute
t0 = time.perf_counter()
filtrado = ddf[(ddf['metodo_pago'] == 'Tarjeta') & (ddf['cantidad'] >= 5)]
resumen_dask = (filtrado.groupby('region')
                .agg({'id_pedido': 'count', 'total': 'sum'})
                .rename(columns={'id_pedido': 'n_ventas', 'total': 'total_facturado'})
                .compute())
t_dask = time.perf_counter() - t0

# Lo mismo con pandas
t0 = time.perf_counter()
df = pd.read_csv(RUTA)
filtrado_pd = df[(df['metodo_pago'] == 'Tarjeta') & (df['cantidad'] >= 5)]
resumen_pandas = (filtrado_pd.groupby('region')
                  .agg(n_ventas=('id_pedido', 'count'), total_facturado=('total', 'sum')))
t_pandas = time.perf_counter() - t0

print(resumen_dask)
print(f'Dask: {t_dask:.3f} s   |   pandas: {t_pandas:.3f} s')
````

````python
import os
import shutil

shutil.rmtree('salida_dask', ignore_errors=True)   # empezamos siempre desde cero

# --- Parte B ---
ddf = dd.read_csv(RUTA, blocksize='256KB')
# 1. Modificar: los descuentos vacíos pasan a 0
ddf['descuento'] = ddf['descuento'].fillna(0)
# 2. Borrar: quitamos las ventas de Barcelona
ddf = ddf[ddf['ciudad'] != 'Barcelona']

# 3. Escribir en paralelo en Parquet (por región) y en CSV (un fichero por partición)
ddf.to_parquet('salida_dask/parquet', partition_on=['region'], write_index=False)
ddf.to_csv('salida_dask/csv/ventas-*.csv', index=False)

for carpeta, _, ficheros in os.walk('salida_dask'):
    if ficheros:
        print(carpeta, '->', len(ficheros), 'ficheros')
````

````python
from concurrent.futures import ThreadPoolExecutor

os.makedirs('salida_dask/json', exist_ok=True)
df = pd.read_csv(RUTA)
grupos = list(df.groupby('categoria'))

def guardar_json(categoria, datos):
    ruta = f'salida_dask/json/{categoria}.json'
    datos.to_json(ruta, orient='records', force_ascii=False, indent=2)
    return ruta

# Secuencial
t0 = time.perf_counter()
for categoria, datos in grupos:
    guardar_json(categoria, datos)
t_secuencial = time.perf_counter() - t0

# En paralelo con 4 hilos
t0 = time.perf_counter()
with ThreadPoolExecutor(max_workers=4) as executor:
    list(executor.map(lambda g: guardar_json(*g), grupos))
t_paralelo = time.perf_counter() - t0

print(f'Secuencial: {t_secuencial:.3f} s   |   Paralelo: {t_paralelo:.3f} s')
````
