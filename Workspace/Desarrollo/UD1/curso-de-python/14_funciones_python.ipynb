# Funciones en Python

Una función agrupa instrucciones con un nombre. Permite reutilizar código, dividir problemas grandes y hacer los programas más fáciles de leer y probar.

## Definir y llamar una función

```python
def nombre_funcion(parametro):
    # instrucciones
    return resultado
```

Los parámetros son datos que recibe la función. `return` entrega un resultado; si no aparece, la función devuelve `None`.

````python
def saludar(nombre):
    return f"Hola, {nombre}!"

mensaje = saludar("Nora")
print(mensaje)
````

## Parámetros y argumentos

Los **parámetros** aparecen en la definición; los **argumentos** son los valores concretos de la llamada. Podemos usar argumentos posicionales o nombrados.

````python
def presentar(nombre, curso):
    return f"{nombre} estudia {curso}."

print(presentar("Leo", "Python"))
print(presentar(curso="IA", nombre="Noa"))
````

## Valores por defecto

Un parámetro puede tener un valor por defecto. Debe ir después de los parámetros obligatorios.

````python
def repetir(texto, veces=2):
    return texto * veces

print(repetir("IA ") )
print(repetir("Python ", 3))
````

## Devolver resultados

Una función puede devolver un valor, varios valores agrupados en una tupla o nada. Devolver resultados facilita probarla y reutilizarla.

````python
def minimo_y_maximo(valores):
    return min(valores), max(valores)

menor, mayor = minimo_y_maximo([8, 3, 11, 5])
print(menor, mayor)

def mostrar_mensaje(texto):
    print(texto)

print(mostrar_mensaje("Esta función devuelve None"))
````

## Ámbito de las variables

Una variable creada dentro de una función es local: solo existe durante esa llamada. Es preferible pasar la información como parámetro y devolver el resultado en lugar de depender de variables globales.

````python
factor = 10

def multiplicar_por_factor(numero):
    factor_local = 2
    return numero * factor_local

print(multiplicar_por_factor(4))
print(factor)
# print(factor_local)  # No existe fuera de la función
````
