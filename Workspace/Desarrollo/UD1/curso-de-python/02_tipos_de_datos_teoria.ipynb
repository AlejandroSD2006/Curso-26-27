# Tipos de datos en Python

Los tipos describen qué clase de información guarda un valor y qué operaciones puede realizar.

# 1. Variables y valores

Una **variable** es un nombre que hace referencia a un valor. Python infiere su tipo automáticamente a partir del valor asignado.

```python
nombre = "Mina"          # str
edad = 20                 # int
altura = 1.68              # float
estudia_ia = True          # bool

print(nombre, edad, altura, estudia_ia)
```

## Tipado dinámico

No declaramos el tipo de una variable antes de usarla. El nombre puede referirse a valores de tipos distintos en momentos distintos, aunque normalmente evitaremos hacerlo porque dificulta leer el programa.

```python
dato = 42
print(dato, type(dato))

dato = "cuarenta y dos"
print(dato, type(dato))
```

# 2. Tipos numéricos

Los números son fundamentales en ciencia de datos e IA: medidas, probabilidades, índices y resultados de modelos.

## `int`, `float` y `complex`

| Tipo | Ejemplo | Uso típico |
|---|---:|---|
| `int` | `42` | contadores, índices, cantidades enteras |
| `float` | `3.14` | medidas, medias, probabilidades |
| `complex` | `2 + 3j` | cálculo científico especializado |

El separador decimal es el punto, no la coma.

```python
muestras = 120
precision = 0.875
impedancia = 3 + 2j

print(type(muestras), type(precision), type(impedancia))
print(muestras / 2)  # La división / devuelve float
```

## Operadores útiles

- `+`, `-`, `*`, `/`: operaciones aritméticas.
- `//`: división entera.
- `%`: resto de una división.
- `**`: potencia.

En IA aparecen con frecuencia al normalizar valores, calcular distancias y recorrer colecciones.

```python
total = 17
lotes = 5
print(total // lotes)  # elementos completos por lote
print(total % lotes)   # elementos restantes
print(2 ** 8)          # 2 elevado a 8
```

# 3. Texto y lógica

Los textos permiten trabajar con etiquetas, documentos y mensajes. Los booleanos representan decisiones: sí/no, verdadero/falso.

## Cadenas: `str`

Una cadena es una secuencia de caracteres delimitada por comillas simples o dobles. Es **inmutable**: sus métodos devuelven una nueva cadena.

```python
texto = "  Python para IA  "
print(texto.upper())
print(texto.strip())
print(texto.replace("IA", "inteligencia artificial"))
print(len(texto))  # len() cuenta elementos
```

## Booleanos: `bool`

Solo existen dos valores booleanos: `True` y `False`. Se obtienen al comparar valores y se usan para controlar el flujo de un programa.

Recuerda: se escriben con mayúscula inicial.

```python
nota = 7.5
aprobado = nota >= 5
tiene_beca = True

print(aprobado)
print(aprobado and tiene_beca)
print(not aprobado)
```

# 4. Colecciones

Una colección agrupa varios valores. Elegir la colección correcta ayuda a expresar claramente la intención del programa.

## Comparativa

| Tipo | Ordenado | Modificable | Duplicados | Acceso típico |
|---|---|---|---|---|
| `list` | Sí | Sí | Sí | posición |
| `tuple` | Sí | No | Sí | posición |
| `set` | No | Sí | No | pertenencia |
| `dict` | Sí* | Sí | claves únicas | clave |

Nota: los diccionarios conservan el orden de inserción en Python moderno.

```python
# Datos de un pequeño experimento
lecturas = [0.8, 0.6, 0.9]                 # list
coordenada = (40.4, -3.7)                 # tuple
etiquetas = {"gato", "perro", "gato"}    # set
modelo = {"nombre": "clasificador", "score": 0.91}  # dict

print(lecturas)
print(coordenada)
print(etiquetas)
print(modelo["score"])
```

## ¿Cuándo usar cada una?

- Usa una **lista** para una secuencia de datos que puede crecer o cambiar.
- Usa una **tupla** para datos agrupados que no deben cambiar, como coordenadas.
- Usa un **conjunto** para eliminar duplicados o comprobar pertenencia.
- Usa un **diccionario** para relacionar claves con valores, como una ficha de un modelo.

# 5. Inspeccionar tipos

Python incorpora herramientas para preguntar qué es un objeto y cómo podemos usarlo.

## `type()`, `isinstance()` y `dir()`

- `type(valor)`: devuelve el tipo exacto del valor.
- `isinstance(valor, tipo)`: comprueba si pertenece a un tipo (recomendado en comprobaciones).
- `dir(valor)`: muestra atributos y métodos disponibles.

En programación orientada a objetos, `isinstance()` también reconoce herencias.

```python
valores = [10, 2.5, "hola", [1, 2]]
for valor in valores:
    print(f"{valor!r:10} -> {type(valor).__name__}")

print(isinstance(10, int))
print(isinstance(10, (int, float)))

metodos_lista = [m for m in dir([]) if not m.startswith("_")]
print(metodos_lista)
```

# 6. Conversión de tipos

Los datos externos suelen llegar como texto: formularios, archivos CSV o APIs. Convertirlos correctamente es un paso esencial.

```python
edad_texto = "21"
peso_texto = "68.5"

edad = int(edad_texto)
peso = float(peso_texto)
mensaje = str(edad) + " años"

print(edad + 1, peso, mensaje)
```

## Conversión segura

Si un texto no representa un número válido, `int()` o `float()` producen un error. Para datos reales conviene validar o capturar la excepción.

La función `convertir_a_entero` es un primer ejemplo de función reutilizable.

```python
def convertir_a_entero(texto):
    """Devuelve un entero o None si el texto no es válido."""
    try:
        return int(texto)
    except ValueError:
        return None

for dato in ["12", "-3", "doce"]:
    print(f"{dato!r} -> {convertir_a_entero(dato)}")
```
