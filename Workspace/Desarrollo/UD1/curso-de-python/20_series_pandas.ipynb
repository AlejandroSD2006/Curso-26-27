# Series en Pandas

Una `Series` es una secuencia unidimensional de valores con etiquetas de índice. Puede imaginarse como una columna de una tabla.

## Crear Series

Se pueden crear desde una lista, una tupla, un diccionario o un valor escalar. Podemos indicar un índice propio y un nombre para la serie.

```python
import pandas as pd

notas = pd.Series([8.5, 7.0, 9.2], name="nota")
temperaturas = pd.Series([18, 20, 19], index=["lunes", "martes", "miércoles"], name="temperatura")
desde_diccionario = pd.Series({"Ada": 8, "Lin": 7, "Noa": 9})

print(notas)
print(temperaturas)
print(desde_diccionario)
```

## Valores, índice y atributos

- `.values` entrega los valores subyacentes.
- `.index` muestra las etiquetas.
- `.name` guarda el nombre descriptivo.
- `.dtype` describe el tipo de datos.
- `.size` y `len()` indican la cantidad de elementos.

```python
print(temperaturas.values)
print(temperaturas.index)
print(temperaturas.name)
print(temperaturas.dtype)
print(temperaturas.size, len(temperaturas))
```

## Selección con `loc` e `iloc`

`loc` selecciona por etiqueta del índice; `iloc` selecciona por posición. Cuando el índice no es numérico, esta distinción resulta especialmente clara.

```python
print(temperaturas.loc["martes"])
print(temperaturas.iloc[1])
print(temperaturas.loc["lunes":"martes"])
print(temperaturas.iloc[0:2])
```

## Series y datos ausentes

Una Series puede contener valores ausentes, normalmente `NaN` en datos numéricos. Pandas ofrece `.isna()` y `.notna()` para detectarlos.

```python
lecturas = pd.Series([10.2, None, 11.0, None], name="lectura")
print(lecturas)
print(lecturas.isna())
print(f"Valores ausentes: {lecturas.isna().sum()}")
```

## Crear Series desde un DataFrame

Seleccionar una sola columna de un `DataFrame` devuelve una Series. Es la conexión más frecuente entre ambas estructuras.

```python
tabla = pd.DataFrame({"nombre": ["Ada", "Lin"], "nota": [8, 7]})
serie_notas = tabla["nota"]
print(serie_notas)
print(type(serie_notas))
```
