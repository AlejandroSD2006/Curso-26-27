# La función `range()`

`range()` genera una secuencia de enteros. Se usa sobre todo junto con `for` cuando conocemos cuántas veces queremos repetir una operación.

## Formas de usarla

- `range(fin)`: desde `0` hasta `fin - 1`.
- `range(inicio, fin)`: desde `inicio` hasta `fin - 1`.
- `range(inicio, fin, paso)`: añade un salto entre valores.

El límite final nunca se incluye.

```python
print(list(range(5)))
print(list(range(3, 8)))
print(list(range(2, 12, 2)))
print(type(range(5)))
```

## Repetir una acción

Si el valor del contador no importa, se suele usar `_` como nombre de variable. Indica que no vamos a utilizarlo.

```python
for _ in range(3):
    print("Procesando una muestra")

for numero in range(1, 6):
    print(f"Iteración {numero}")
```

## Pasos negativos

Un paso negativo permite contar hacia atrás. En este caso, el inicio debe ser mayor que el final.

```python
for segundos in range(5, 0, -1):
    print(segundos)
print("¡Comienza!")

print(list(range(10, -1, -2)))
```

## `range()` e índices

Para recorrer elementos se prefiere `for elemento in lista`. Usa `range(len(lista))` solo si realmente necesitas acceder o modificar mediante índices.

```python
mediciones = [18.2, 19.0, 18.7]
for indice in range(len(mediciones)):
    print(f"Medición {indice}: {mediciones[indice]}")

# Alternativa preferida si no necesitamos el índice
for medicion in mediciones:
    print(medicion)
```
