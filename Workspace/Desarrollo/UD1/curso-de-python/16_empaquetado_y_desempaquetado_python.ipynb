# Empaquetado y desempaquetado en Python

## Listas, tuplas, diccionarios, `*`, `_` e intercambio de variables

Python permite agrupar varios valores y asignarlos de forma clara a varias variables. Estas técnicas hacen el código más expresivo y aparecen con frecuencia en funciones y recorridos de datos.

# 1. Empaquetado

**Empaquetar** consiste en reunir varios valores en una misma estructura. Al separar valores con comas, Python los empaqueta automáticamente en una tupla.

```python
coordenadas = 37.389, -5.984
colores = "rojo", "verde", "azul"

print(coordenadas)
print(type(coordenadas))
print(colores)

# Los paréntesis son opcionales al crear una tupla
otra_tupla = (1, 2, 3)
print(otra_tupla)
```

## Empaquetar en colecciones

Podemos decidir explícitamente el tipo de colección. Las listas son mutables, las tuplas son inmutables y los diccionarios relacionan claves con valores.

```python
nombre, edad, activo = "Ada", 28, True

datos_lista = [nombre, edad, activo]
datos_tupla = (nombre, edad, activo)
datos_diccionario = {"nombre": nombre, "edad": edad, "activo": activo}

print(datos_lista)
print(datos_tupla)
print(datos_diccionario)
```

# 2. Desempaquetar secuencias

**Desempaquetar** asigna los elementos de una secuencia a varias variables. El número de variables debe coincidir con el número de elementos, salvo que se use `*`.

```python
punto = (10, 20)
x, y = punto
print(f"x = {x}, y = {y}")

materias = ["Python", "Datos", "IA"]
primera, segunda, tercera = materias
print(primera, segunda, tercera)
```

## Desempaquetar con `*`

El asterisco recoge varios elementos restantes en una **lista**. Puede usarse una vez en la asignación y situarse al principio, en medio o al final.

```python
numeros = [10, 20, 30, 40, 50]

primero, *resto = numeros
print(primero, resto)

*inicio, ultimo = numeros
print(inicio, ultimo)

primero, *centro, ultimo = numeros
print(primero, centro, ultimo)
print(type(centro))
```

## Un caso útil: cabecera y datos

Al procesar una fila de datos, a menudo conocemos el primer o el último campo, pero no cuántos valores intermedios habrá. `*` permite expresar esta idea directamente.

```python
registro = ["sensor-01", 18.4, 18.9, 19.1, "correcto"]
sensor, *mediciones, estado = registro

print(f"Sensor: {sensor}")
print(f"Mediciones: {mediciones}")
print(f"Estado: {estado}")
print(f"Media: {sum(mediciones) / len(mediciones):.2f}")
```

# 3. El uso de `_`

Por convención, `_` significa que un valor existe pero no nos interesa utilizarlo. No es una palabra especial: sigue siendo un nombre de variable válido.

```python
# Ignoramos el segundo dato porque no lo necesitamos
nombre, _, ciudad = ("Nora", 23, "Sevilla")
print(nombre, ciudad)

# Repetimos una acción sin usar el contador
for _ in range(3):
    print("Guardando una copia de seguridad")
```

## `_` y valores restantes

Podemos combinar `_` con `*` si queremos ignorar varios valores. En este ejemplo nos quedamos solo con el primero y el último.

Aunque `*_` funciona, un nombre como `*ignorados` suele explicar mejor el código cuando es complejo.

```python
datos = ["inicio", 1, 2, 3, "fin"]
primero, *_, ultimo = datos
print(primero, ultimo)

anio, mes, _ = (2026, 9, 12)
print(f"{mes:02}/{anio}")
```

# 4. Intercambiar variables

El desempaquetado permite intercambiar dos variables sin crear una variable temporal. Python empaqueta los valores del lado derecho y los desempaqueta a la izquierda.

```python
a = 3
b = 7
print(f"Antes: a={a}, b={b}")

a, b = b, a
print(f"Después: a={a}, b={b}")

# También se puede hacer una rotación de tres valores
primero, segundo, tercero = "A", "B", "C"
primero, segundo, tercero = tercero, primero, segundo
print(primero, segundo, tercero)
```

# 5. Diccionarios y `**`

Al iterar un diccionario directamente se obtienen las claves. Con `.items()` obtenemos pares clave-valor, que se pueden desempaquetar.

El operador `**` desempaqueta un diccionario en pares `clave=valor`.

```python
contacto = {"nombre": "Amir", "telefono": "600 111 222", "ciudad": "Córdoba"}

for clave, valor in contacto.items():
    print(f"{clave.title()}: {valor}")

claves = list(contacto)
print(claves)
```

## Combinar y copiar diccionarios

`{**diccionario1, **diccionario2}` crea un diccionario nuevo. Si hay claves repetidas, prevalece el valor que aparece más a la derecha.

```python
datos_basicos = {"nombre": "Ada", "curso": "Python"}
datos_extra = {"curso": "Python para IA", "activo": True}
perfil = {**datos_basicos, **datos_extra}

print(perfil)
copia = {**perfil}
print(copia is perfil)  # Son diccionarios distintos
```

# 6. `*args` y `**kwargs` en funciones

En la definición de una función:

- `*args` empaqueta argumentos posicionales extra en una tupla.
- `**kwargs` empaqueta argumentos nombrados extra en un diccionario.

Los nombres `args` y `kwargs` son convencionales; los asteriscos son lo importante.

```python
def sumar(*numeros):
    return sum(numeros)

def mostrar_ficha(**datos):
    for clave, valor in datos.items():
        print(f"{clave.title()}: {valor}")

print(sumar(3, 5, 8, 10))
mostrar_ficha(nombre="Lía", curso="IA", nivel=1)
```

## Desempaquetar al llamar una función

Los operadores `*` y `**` también desempaquetan valores al realizar una llamada. La secuencia proporciona argumentos posicionales; el diccionario proporciona argumentos nombrados.

```python
def presentar(nombre, curso, nivel):
    return f"{nombre} cursa {curso}, nivel {nivel}."

argumentos = ["Sara", "Python", 2]
print(presentar(*argumentos))

datos = {"nombre": "Omar", "curso": "IA", "nivel": 1}
print(presentar(**datos))
```
