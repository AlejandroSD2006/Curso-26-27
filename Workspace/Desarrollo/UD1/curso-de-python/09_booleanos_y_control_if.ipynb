# Booleanos y control de flujo

Los booleanos representan decisiones y la estructura `if` permite que un programa elija qué instrucciones ejecutar.

## El tipo `bool`

Solo tiene dos valores: `True` y `False`. Se obtienen con comparaciones y se combinan con operadores lógicos.

Las palabras se escriben con mayúscula inicial.

```python
edad = 18
tiene_permiso = True

print(edad >= 18)
print(type(tiene_permiso))
print(edad >= 18 and tiene_permiso)
```

## Operadores de comparación

| Operador | Significado |
|---|---|
| `==` | igual que |
| `!=` | distinto de |
| `>`, `<` | mayor o menor que |
| `>=`, `<=` | mayor o menor, incluyendo igualdad |
| `in` | pertenece a una colección |

```python
nota = 7
etiquetas = ["gato", "perro", "ave"]
print(nota == 7)
print(nota != 5)
print("perro" in etiquetas)
print("pez" not in etiquetas)
```

## Operadores lógicos

- `and`: ambas condiciones deben ser verdaderas.
- `or`: basta con que una condición sea verdadera.
- `not`: invierte un booleano.

Usa paréntesis si una condición compleja pierde claridad.

```python
edad = 17
autorizacion = True
es_fin_de_semana = False

print(edad >= 18 or autorizacion)
print(edad >= 18 and autorizacion)
print(not es_fin_de_semana)
```

# La estructura `if`

`if` ejecuta un bloque solo si su condición es verdadera. El sangrado de cuatro espacios forma parte de la sintaxis de Python.

```python
precision = 0.91
if precision >= 0.90:
    print("El modelo supera el objetivo de precisión.")

print("Fin de la evaluación.")
```

## `if` y `else`

`else` contiene el camino alternativo: se ejecuta cuando la condición de `if` es falsa.

```python
temperatura = 26
if temperatura > 25:
    print("Hace calor")
else:
    print("La temperatura es agradable")
```

## `if` - `elif` - `else`

`elif` permite evaluar alternativas en orden. Python ejecuta solo el primer bloque cuya condición sea verdadera y después abandona la cadena.

```python
nota = 8.2
if nota >= 9:
    calificacion = "Sobresaliente"
elif nota >= 7:
    calificacion = "Notable"
elif nota >= 5:
    calificacion = "Aprobado"
else:
    calificacion = "Suspenso"

print(calificacion)
```

## Valores considerados falsos

En una condición también se interpretan como falsos: `False`, `None`, `0`, `0.0`, `""`, `[]`, `{}` y `()`. El resto de valores suele interpretarse como verdadero.

Este comportamiento permite comprobar rápidamente si una colección o cadena contiene algo.

```python
mensajes = []
if mensajes:
    print("Hay mensajes pendientes")
else:
    print("No hay mensajes pendientes")

nombre = "Ada"
if nombre:
    print(f"Nombre recibido: {nombre}")
```
