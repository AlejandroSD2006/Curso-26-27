# El bucle `for` en Python

`for` permite repetir un bloque de código para cada elemento de una colección. Es la forma habitual de recorrer listas, cadenas, tuplas, conjuntos y diccionarios.

## Sintaxis

```python
for elemento in coleccion:
    # instrucciones que se repiten
```

La variable `elemento` toma sucesivamente cada valor de la colección. El sangrado de cuatro espacios delimita el bloque.

````python
lenguajes = ["Python", "R", "Julia"]
for lenguaje in lenguajes:
    print(f"Lenguaje: {lenguaje}")
````

## Recorrer tipos de colecciones

Una cadena se recorre carácter a carácter. En un diccionario, el recorrido directo produce sus claves; con `.items()` obtenemos claves y valores.

````python
for letra in "IA":
    print(letra)

metricas = {"accuracy": 0.91, "recall": 0.87}
for nombre, valor in metricas.items():
    print(f"{nombre}: {valor:.1%}")
````

## Índice y valor: `enumerate()`

Cuando necesitas la posición y el valor, usa `enumerate()`. Es más claro que crear un contador manual.

````python
tareas = ["cargar datos", "entrenar", "evaluar"]
for numero, tarea in enumerate(tareas, start=1):
    print(f"Paso {numero}: {tarea}")
````

## Acumuladores y contadores

Un acumulador guarda un resultado que se actualiza en cada vuelta. Un contador registra cuántas veces ocurre algo. Inicialízalos antes del bucle.

````python
notas = [8, 5, 9, 4, 7]
suma = 0
aprobadas = 0

for nota in notas:
    suma += nota
    if nota >= 5:
        aprobadas += 1

print(f"Media: {suma / len(notas):.1f}")
print(f"Aprobadas: {aprobadas}")
````

## `break` y `continue`

- `break` termina el bucle inmediatamente.
- `continue` salta a la siguiente vuelta.

Úsalos con intención: un bucle sencillo suele ser más fácil de entender.

````python
datos = [12, None, 8, -1, 15]
for dato in datos:
    if dato is None:
        continue
    if dato < 0:
        print("Dato negativo: se detiene el análisis")
        break
    print(f"Dato válido: {dato}")
````
