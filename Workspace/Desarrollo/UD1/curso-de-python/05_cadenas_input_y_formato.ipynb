# 1. El tipo `str`

Una cadena es una secuencia ordenada de caracteres. Puede escribirse con comillas simples, dobles o triples para texto de varias líneas.

````python
saludo = "Hola"
asignatura = 'Python para IA'
mensaje_largo = """Este texto
ocupa varias líneas."""

print(saludo)
print(asignatura)
print(mensaje_largo)
print(type(asignatura))
````

## Índices y cortes

Cada carácter tiene una posición o índice. El primer índice es `0`; el último se puede pedir con `-1`. Los cortes usan la forma `texto[inicio:fin]`, sin incluir la posición final.

````python
palabra = "Python"
print(palabra[0])     # P
print(palabra[-1])    # n
print(palabra[0:3])   # Pyt
print(palabra[::2])   # Pto
print(len(palabra))
````

## Las cadenas son inmutables

No podemos cambiar directamente un carácter de una cadena. En su lugar, los métodos crean y devuelven una cadena nueva.

````python
texto = "python"
# texto[0] = "P"  # Produciría TypeError
texto_con_mayuscula = texto.capitalize()
print(texto)
print(texto_con_mayuscula)
````

# 2. Métodos fundamentales de cadenas

Un método se aplica a un objeto con la forma `cadena.metodo()`.

## Cambiar mayúsculas y limpiar

- `.lower()` y `.upper()` cambian el uso de mayúsculas.
- `.capitalize()` pone en mayúscula el primer carácter.
- `.title()` pone en mayúscula el inicio de cada palabra.
- `.strip()` elimina espacios al principio y final.
- `.replace(viejo, nuevo)` sustituye texto.

````python
entrada = "  python Y datos  "
print(entrada.lower())
print(entrada.upper())
print(entrada.strip().title())
print(entrada.strip().replace("datos", "IA"))
````

## Buscar y dividir texto

- `.startswith()` y `.endswith()` comprueban comienzos y finales.
- `.find()` devuelve la posición de un texto, o `-1` si no aparece.
- `.count()` cuenta apariciones.
- `.split()` separa una cadena en una lista.
- `separador.join(lista)` une elementos de texto.

````python
frase = "Python permite crear modelos de IA"
print(frase.startswith("Python"))
print(frase.find("modelos"))
print(frase.count("a"))
palabras = frase.split()
print(palabras)
print("-".join(palabras))
````

## Comprobar el contenido

Los métodos `is...` devuelven un booleano. Son útiles para validar datos que llegan en forma de texto.

Prueba `isdigit()`, `isalpha()`, `isalnum()` y `isspace()`.

````python
muestras = ["123", "doce", "A12", "   "]
for muestra in muestras:
    print(f"{muestra!r:8} dígitos={muestra.isdigit()} letras={muestra.isalpha()} alfanumérico={muestra.isalnum()}")
````

# 3. Entrada con `input()`

`input(mensaje)` muestra un mensaje y espera que la persona escriba una respuesta. **Siempre devuelve una cadena**, aunque se introduzcan números.

````python
# Ejecuta esta celda e introduce tu nombre.
nombre = input("¿Cómo te llamas? " )
print(f"Hola, {nombre}!")
print(f"El tipo recibido es: {type(nombre).__name__}")
````

## Convertir datos introducidos

Para realizar operaciones numéricas, convierte la respuesta: `int()` para enteros y `float()` para decimales.

No conviertas sin pensar: si la persona escribe texto no numérico, se producirá un `ValueError`.

````python
# Ejecuta esta celda y escribe una edad entera.
edad_texto = input("¿Qué edad tienes? " )
edad = int(edad_texto)
print(f"El próximo año tendrás {edad + 1} años.")
````

## Una función para pedir enteros

Las funciones permiten reutilizar una solución. Esta pide un entero y vuelve a preguntar si la entrada no es válida.

Se introduce también el bucle `while`, que repite instrucciones mientras la condición sea verdadera.

````python
def pedir_entero(mensaje):
    """Solicita un entero hasta recibir una entrada válida."""
    while True:
        texto = input(mensaje).strip()
        try:
            return int(texto)
        except ValueError:
            print("Escribe un número entero válido.")

# Descomenta para probar la función de forma interactiva:
# numero_de_muestras = pedir_entero("Número de muestras: " )
# print(numero_de_muestras)
````

# 4. Formatear cadenas

Formatear es insertar valores en un texto de forma clara. En Python actual, las **f-strings** suelen ser la opción más cómoda y legible.

## f-strings

Anteponemos `f` a la cadena y escribimos expresiones entre llaves: `f"Resultado: {valor}"`. Dentro de las llaves se puede usar una variable, una operación o una llamada sencilla a función.

````python
nombre = "Amira"
aciertos = 18
total = 20

print(f"{nombre} obtuvo {aciertos} de {total} aciertos.")
print(f"Porcentaje: {aciertos / total * 100}%")
print(f"Nombre normalizado: {nombre.upper()}")
````

## Decimales, anchura y alineación

Después de `:` podemos indicar cómo mostrar el valor:

- `.2f`: dos decimales.
- `>10`, `<10`, `^10`: alinear en un campo de 10 caracteres.
- `,`: separador de miles.
- `.1%`: porcentaje con un decimal.

````python
precision = 0.93456
usuarios = 12500
modelo = "Clasificador"

print(f"Precisión: {precision:.2%}")
print(f"Usuarios: {usuarios:,}")
print(f"|{modelo:<15}|{precision:>8.3f}|")
print(f"|{'RESULTADO':^15}|{'OK':^8}|")
````

## Otras formas de formato

También existen `.format()` y el operador `%`, sobre todo en código anterior. Conviene reconocerlos, pero en código nuevo prefiere f-strings cuando sea posible.

````python
lenguaje = "Python"
version = 3
print("{} es el lenguaje del curso (versión {}).".format(lenguaje, version))
print("%s es el lenguaje del curso (versión %d)." % (lenguaje, version))
print(f"{lenguaje} es el lenguaje del curso (versión {version}).")
````
