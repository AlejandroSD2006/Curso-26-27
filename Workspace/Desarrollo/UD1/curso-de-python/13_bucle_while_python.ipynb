# El bucle `while` en Python

`while` repite un bloque mientras una condición sea verdadera. Es útil cuando no sabemos de antemano cuántas iteraciones habrá.

## Sintaxis y precaución

```python
while condicion:
    # instrucciones
```

Dentro del bucle debe ocurrir algo que pueda volver falsa la condición. De lo contrario tendremos un bucle infinito.

````python
contador = 1
while contador <= 5:
    print(contador)
    contador += 1

print("Fin")
````

## Acumuladores con `while`

Podemos usar un contador y un acumulador para construir resultados de forma repetida.

````python
numero = 1
suma = 0
while numero <= 10:
    suma += numero
    numero += 1
print(f"Suma: {suma}")
````

## Validar una entrada

`while` permite repetir una pregunta hasta que la persona proporcione un dato válido. Aquí se muestra una función; descomenta la llamada si quieres probarla de forma interactiva.

````python
def pedir_positivo(mensaje):
    while True:
        try:
            numero = int(input(mensaje))
            if numero > 0:
                return numero
            print("Debe ser mayor que cero.")
        except ValueError:
            print("Escribe un número entero.")

# cantidad = pedir_positivo("Cantidad: " )
````

## `break`, `continue` y bandera booleana

`break` termina el bucle. `continue` salta una vuelta. Una **bandera** es una variable booleana que expresa si el bucle debe continuar.

````python
datos = [4, 0, -2, 8]
indice = 0
en_proceso = True

while en_proceso and indice < len(datos):
    dato = datos[indice]
    indice += 1
    if dato == 0:
        continue
    if dato < 0:
        print("Dato negativo: fin del proceso")
        en_proceso = False
    else:
        print(f"Procesado: {dato}")
````
