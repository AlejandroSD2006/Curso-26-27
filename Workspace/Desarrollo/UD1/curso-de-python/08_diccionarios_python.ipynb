# Diccionarios en Python

Un diccionario relaciona **claves** únicas con valores. Es ideal para representar fichas, configuraciones y datos con nombre.

## Características

- Se crean con llaves: `{clave: valor}`.
- Las claves deben ser únicas e inmutables, normalmente cadenas o números.
- Los valores pueden ser de cualquier tipo.
- Son mutables y conservan el orden de inserción en Python moderno.
- El acceso se realiza por clave, no por posición.

```python
modelo = {
    "nombre": "Clasificador de imágenes",
    "precision": 0.93,
    "activo": True
}
print(modelo)
print(modelo["nombre"])
print(type(modelo))
```

## Añadir, modificar y eliminar

Asignar a una clave nueva añade un par. Asignar a una clave existente actualiza su valor. `del` elimina una clave; `.pop()` la elimina y devuelve el valor.

```python
perfil = {"nombre": "Ana", "curso": "IA"}
perfil["edad"] = 22
perfil["curso"] = "Python para IA"
edad = perfil.pop("edad")
print(perfil)
print(f"Edad extraída: {edad}")
```

## Acceder de forma segura

`diccionario[clave]` produce `KeyError` si no existe la clave. `.get(clave, valor_por_defecto)` evita ese error cuando el dato puede faltar.

```python
persona = {"nombre": "Ana", "edad": 30}
print(persona.get("apellido", "No especificado"))
print("edad" in persona)
print("ciudad" in persona)
```

## Recorrer un diccionario

- `.keys()` obtiene las claves.
- `.values()` obtiene los valores.
- `.items()` entrega pares `(clave, valor)` y es muy útil con `for`.

También se puede usar `len()` para saber el número de pares.

```python
metricas = {"accuracy": 0.91, "precision": 0.88, "recall": 0.86}
for nombre, valor in metricas.items():
    print(f"{nombre.title():<12} {valor:.1%}")
print(f"Total de métricas: {len(metricas)}")
```

## Diccionarios anidados

Un valor puede ser otra colección. Esto permite describir estructuras más completas, por ejemplo un modelo y sus resultados.

```python
experimento = {
    "modelo": "Regresión logística",
    "resultado": {"accuracy": 0.89, "muestras": 1200},
    "etiquetas": ["positivo", "negativo"]
}
print(experimento["resultado"]["accuracy"])
print(experimento["etiquetas"][0])
```
