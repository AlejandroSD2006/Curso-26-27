# Filtrado de Series en Pandas

Filtrar consiste en seleccionar solo los elementos que cumplen una condición. En Pandas se crea una máscara booleana y se utiliza para seleccionar valores.

## Máscaras booleanas

Una comparación sobre una Series devuelve otra Series de `True` y `False`. Al usar esa máscara entre corchetes obtenemos únicamente los valores donde es `True`.

```python
import pandas as pd

notas = pd.Series([8.5, 4.0, 7.2, 9.0, 5.0], index=["Ada", "Lin", "Noa", "Omar", "Vera"])
aprobados = notas >= 5
print(aprobados)
print(notas[aprobados])
print(notas[notas >= 5])
```

## Combinar condiciones

Usa `&` para **y**, `|` para **o** y `~` para **no**. Cada condición debe ir entre paréntesis. No uses `and` u `or` con Series: están pensados para booleanos individuales.

```python
print(notas[(notas >= 5) & (notas < 9)])
print(notas[(notas < 5) | (notas == 9)])
print(notas[~(notas >= 5)])
```

## Métodos de filtrado

- `.between(inicio, fin)`: intervalo inclusivo.
- `.isin(valores)`: pertenencia a una colección.
- `.isna()` y `.notna()`: valores ausentes.
- `.str.contains(patron)`: texto que contiene un patrón.

```python
print(notas[notas.between(6, 8.5)])
print(notas[notas.isin([4.0, 9.0])])

ciudades = pd.Series(["Sevilla", "Córdoba", "Salamanca", "Madrid"], dtype="string")
print(ciudades[ciudades.str.contains("a", case=False)])
print(ciudades[ciudades.str.startswith("S")])
```

## Filtrar y asignar con cuidado

Para modificar valores que cumplen una condición, usa `.loc[mascara]`. Evita asignaciones encadenadas que puedan producir avisos y resultados ambiguos.

```python
calificaciones = notas.copy()
calificaciones.loc[calificaciones < 5] = 5
print(calificaciones)

estado = pd.Series("pendiente", index=notas.index)
estado.loc[notas >= 5] = "aprobado"
print(estado)
```

## La función `where()`

`serie.where(condicion, otro)` conserva los valores que cumplen la condición y sustituye el resto por `otro`. Es útil para crear versiones transformadas sin perder el índice.

```python
resultado = notas.where(notas >= 5, "suspenso")
print(resultado)

etiquetas = notas.where(notas >= 8, "por mejorar")
print(etiquetas)
```
