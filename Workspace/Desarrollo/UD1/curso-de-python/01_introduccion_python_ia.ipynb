# 1. ¿Qué es Python?

Python es un lenguaje de programación de propósito general, interpretado y de alto nivel.

Su filosofía prioriza que el código sea claro, legible y fácil de mantener.

## Características principales

- **Sintaxis legible:** usa sangrado para delimitar bloques.
- **Interpretado:** se ejecuta sin una compilación manual previa.
- **Tipado dinámico:** el tipo se asocia al valor durante la ejecución.
- **Multiparadigma:** permite programación imperativa, funcional y orientada a objetos.
- **Multiplataforma:** el mismo código suele funcionar en Windows, macOS y Linux.
- **Código abierto:** dispone de una comunidad global muy activa.

```python
# Un ejemplo de código claro y breve
nombre = "Ada"
saludo = f"Hola, {nombre}. ¡Bienvenida a Python!"
print(saludo)
```

## La legibilidad importa

Python busca que el programa se parezca a una explicación de la solución. Esto facilita colaborar, revisar código y aprender.

> *"La legibilidad cuenta"* es uno de los principios del Zen de Python.

```python
# El bloque de instrucciones se marca con sangrado
temperatura = 28

if temperatura > 25:
    print("Hace calor")
else:
    print("La temperatura es agradable")
```

# 5. Tres funciones para explorar

Cuando aprendemos, Python puede describir sus propios objetos. `type()`, `dir()` y `help()` son herramientas esenciales para investigar.

## `type()` — ¿qué clase de valor es?

Devuelve el tipo de un objeto. Es muy útil para comprobar qué dato hemos creado o recibido de una función.

```python
edad = 18
nombre = "Lucía"
notas = [8.5, 7.0, 9.0]

print(type(edad))
print(type(nombre))
print(type(notas))
```

## `dir()` — ¿qué puedo hacer con este objeto?

Devuelve los nombres de atributos y métodos disponibles. Para empezar, fíjate especialmente en los que no comienzan por guion bajo.

```python
texto = "inteligencia artificial"
metodos_publicos = [nombre for nombre in dir(texto) if not nombre.startswith("_")]
print(metodos_publicos)
```

## `help()` — ¿cómo se utiliza?

Muestra la documentación integrada de un objeto, función, clase o método. En un cuaderno, su salida aparece debajo de la celda.

La documentación es parte de la herramienta de trabajo: aprender a leerla te hará más autónomo.

```python
# Ejecuta esta celda y busca qué argumentos acepta sorted
help(sorted)
```
