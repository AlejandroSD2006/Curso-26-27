# Operaciones básicas con Series

Pandas aplica muchas operaciones de forma vectorizada: una sola expresión opera sobre todos los elementos de una `Series`, sin escribir un bucle explícito.

## Operaciones aritméticas

Podemos sumar, restar, multiplicar, dividir y elevar Series. Cuando combinamos Series, Pandas alinea los valores por sus etiquetas de índice.

```python
import pandas as pd

ventas = pd.Series([120, 150, 100], index=["enero", "febrero", "marzo"])
print(ventas + 10)
print(ventas * 1.21)
print(ventas / 2)
print(ventas ** 2)
```

## Alineación por índice

Al sumar Series con índices distintos, Pandas hace coincidir etiquetas. Si una etiqueta falta en una Series, el resultado para ella será `NaN`.

```python
tienda_a = pd.Series({"lunes": 10, "martes": 12, "miércoles": 9})
tienda_b = pd.Series({"martes": 8, "miércoles": 11, "jueves": 7})

print(tienda_a + tienda_b)
print(tienda_a.add(tienda_b, fill_value=0))
```

## Comparaciones

Las comparaciones devuelven una Series booleana. Es la base para el filtrado de datos.

```python
notas = pd.Series([8.5, 4.0, 7.2, 9.0, 5.0])
print(notas >= 5)
print(notas == 9.0)
print(notas.between(5, 8))
```

## Métodos de texto: `.str`

Para aplicar métodos de cadenas a todos los elementos de una Series de texto se usa el accesor `.str`. Por ejemplo: `.str.lower()`, `.str.contains()` y `.str.len()`.

```python
nombres = pd.Series([" Ada ", "LIN", "Noa"], dtype="string")
limpios = nombres.str.strip().str.title()
print(limpios)
print(limpios.str.len())
print(limpios.str.contains("a", case=False))
```

## Ordenar y valores únicos

- `.sort_values()` ordena según los valores.
- `.sort_index()` ordena por índice.
- `.unique()` obtiene los valores distintos.
- `.value_counts()` cuenta frecuencias.

```python
etiquetas = pd.Series(["gato", "perro", "gato", "ave", "perro", "gato"])
print(etiquetas.sort_values().tolist())
print(etiquetas.unique())
print(etiquetas.value_counts())
```
