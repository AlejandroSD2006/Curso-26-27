# Limpieza de datos con Pandas

Los datos reales suelen tener valores ausentes, texto con formatos inconsistentes, duplicados y tipos incorrectos. Limpiar no es borrar sin pensar: primero inspeccionamos, después decidimos.

## Datos ausentes

Pandas representa ausencias numéricas con `NaN` y también reconoce `None` o `pd.NA`.

- `.isna()` detecta ausencias.
- `.dropna()` elimina filas o valores ausentes.
- `.fillna(valor)` los reemplaza.

La mejor decisión depende del significado de los datos.

```python
import pandas as pd

datos = pd.DataFrame({
    "nombre": ["Ada", "Lin", None, "Noa"],
    "nota": [8.5, None, 7.0, None],
    "curso": ["IA", "IA", "Python", "Python"]
})
display(datos)
display(datos.isna())
print(datos.isna().sum())
```

```python
sin_ausentes = datos.dropna()
notas_con_cero = datos.copy()
notas_con_cero["nota"] = notas_con_cero["nota"].fillna(0)
nombres_desconocidos = datos.fillna({"nombre": "Sin nombre"})

display(sin_ausentes)
display(notas_con_cero)
display(nombres_desconocidos)
```

## Duplicados

`.duplicated()` identifica filas repetidas y `.drop_duplicates()` las elimina. Antes de eliminar, decide qué columnas definen un duplicado real.

```python
inscripciones = pd.DataFrame({
    "nombre": ["Ada", "Lin", "Ada", "Noa"],
    "curso": ["IA", "Python", "IA", "Python"]
})
print(inscripciones.duplicated())
display(inscripciones.drop_duplicates())
display(inscripciones.drop_duplicates(subset=["nombre"]))
```

## Limpiar texto

El accesor `.str` aplica operaciones a toda una columna de texto. Acciones habituales: eliminar espacios, normalizar mayúsculas y reemplazar variantes de un mismo valor.

```python
ciudades = pd.Series([" sevilla", "SEVILLA ", "Córdoba", "cordoba"], dtype="string")
limpias = ciudades.str.strip().str.title().str.normalize("NFKD").str.encode("ascii", "ignore").str.decode("ascii")
print(limpias)
print(limpias.replace({"Cordoba": "Córdoba"}))
```

## Convertir tipos de forma robusta

`pd.to_numeric(..., errors="coerce")` transforma en `NaN` los valores no convertibles. Después podemos detectar y decidir qué hacer con ellos.

```python
precios_texto = pd.Series(["10.5", "12", "sin dato", "8.75"])
precios = pd.to_numeric(precios_texto, errors="coerce")
print(precios)
print(precios.isna())
print(precios.fillna(precios.median()))
```

## Renombrar columnas

Nombres consistentes simplifican el análisis. `.rename()` permite corregirlos; una convención habitual es usar minúsculas y guion bajo.

```python
tabla = pd.DataFrame({"Nombre Completo": ["Ada"], "Nota Final": [9]})
tabla = tabla.rename(columns={"Nombre Completo": "nombre_completo", "Nota Final": "nota_final"})
display(tabla)
```
