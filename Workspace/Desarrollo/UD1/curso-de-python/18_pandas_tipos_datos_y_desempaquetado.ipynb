# Tipos de datos en Pandas y desempaquetado

Pandas transforma datos de Python en estructuras con tipos especializados. Comprender los tipos evita errores al calcular, limpiar y filtrar.

## Tipos habituales en Pandas

- `int64`: enteros.
- `float64`: números decimales.
- `object` o `string`: texto u objetos generales.
- `bool` o `boolean`: valores lógicos.
- `datetime64[ns]`: fechas y horas.
- `category`: categorías repetidas y limitadas.

El atributo `.dtype` describe una `Series`; `.dtypes` describe las columnas de un `DataFrame`.

```python
import pandas as pd

valores = pd.Series([10, 20, 30])
precios = pd.Series([2.5, 3.1, 4.0])
nombres = pd.Series(["Ada", "Lin", "Noa"], dtype="string")
activo = pd.Series([True, False, True], dtype="boolean")

print(valores.dtype, precios.dtype, nombres.dtype, activo.dtype)
```

## Convertir tipos: `astype()`

Los datos externos pueden venir con tipos inadecuados. `astype()` crea una versión con el tipo indicado. Convierte solo cuando tenga sentido: un identificador como `"0012"` suele ser texto, no un número para operar.

```python
datos = pd.Series(["10", "20", "30"], name="cantidad")
print(datos.dtype)
cantidades = datos.astype("int64")
print(cantidades.dtype)
print(cantidades + 5)
```

## Fechas y categorías

`pd.to_datetime()` convierte texto en fechas. El tipo `category` es útil cuando una columna solo admite unas pocas etiquetas repetidas, como nivel o estado.

```python
fechas = pd.to_datetime(pd.Series(["2026-09-01", "2026-09-15"]))
nivel = pd.Series(["inicial", "avanzado", "inicial"], dtype="category")
print(fechas.dtype)
print(nivel.dtype)
print(nivel.cat.categories)
```

# Desempaquetado de Python

El desempaquetado asigna varios valores de una secuencia a varias variables. Pandas no lo reemplaza: estas técnicas de Python siguen siendo útiles al construir y recorrer datos.

```python
registro = ("Ana", 8.5, "Aprobado")
nombre, nota, estado = registro
print(nombre, nota, estado)

primero, *centrales, ultimo = [1, 2, 3, 4, 5]
print(primero, centrales, ultimo)

# _ expresa que no necesitamos un valor
anio, mes, _ = (2026, 9, 12)
print(anio, mes)
```

## Desempaquetar diccionarios y filas

Al iterar `diccionario.items()` obtenemos pares clave-valor. Al crear un `DataFrame` desde una lista de diccionarios, cada diccionario puede representar una fila.

```python
fila = {"nombre": "Ada", "nota": 9}
for clave, valor in fila.items():
    print(f"{clave}: {valor}")

filas = [
    {"nombre": "Ada", "nota": 9},
    {"nombre": "Lin", "nota": 7}
]
tabla = pd.DataFrame(filas)
display(tabla)
print(tabla.dtypes)
```
