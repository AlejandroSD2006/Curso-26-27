# Funciones que llaman a funciones

Los programas crecen mejor cuando dividimos una tarea en funciones pequeñas. Una función principal puede coordinar otras funciones especializadas.

## Descomposición de un problema

En lugar de escribir una función enorme, separamos responsabilidades:

- una función valida datos;
- otra calcula un resultado;
- otra lo presenta;
- una función principal las llama en orden.

Esto favorece la reutilización y la comprobación de cada parte.

```python
def elevar_al_cuadrado(numero):
    return numero ** 2

def mostrar_cuadrado(numero):
    resultado = elevar_al_cuadrado(numero)
    print(f"El cuadrado de {numero} es {resultado}.")

mostrar_cuadrado(6)
```

## Ejemplo: analizar una lista de notas

Cada función realiza una única tarea. `informe_notas` coordina las demás y no repite su lógica.

```python
def calcular_media(notas):
    return sum(notas) / len(notas)

def contar_aprobadas(notas):
    return sum(nota >= 5 for nota in notas)

def informe_notas(notas):
    media = calcular_media(notas)
    aprobadas = contar_aprobadas(notas)
    return f"Media: {media:.1f}; aprobadas: {aprobadas}/{len(notas)}"

print(informe_notas([8, 4, 7, 9, 5]))
```

## Orden de llamadas y retorno

Cuando una función llama a otra, Python termina la función interna, recoge su `return` y vuelve a la línea siguiente de la función que la llamó.

Un `print()` muestra información; `return` entrega un valor que puede usar otra función.

```python
def limpiar_nombre(nombre):
    return nombre.strip().title()

def crear_saludo(nombre):
    nombre_limpio = limpiar_nombre(nombre)
    return f"Bienvenida, {nombre_limpio}."

def imprimir_saludo(nombre):
    print(crear_saludo(nombre))

imprimir_saludo("  aDA  ")
```

## Ejemplo: validar, calcular y presentar

Este patrón aparece constantemente en aplicaciones: comprobar primero, procesar después y mostrar al final.

```python
def lista_numerica(valores):
    return bool(valores) and all(isinstance(valor, (int, float)) for valor in valores)

def calcular_media_segura(valores):
    if not lista_numerica(valores):
        return None
    return sum(valores) / len(valores)

def mostrar_media(valores):
    media = calcular_media_segura(valores)
    if media is None:
        print("No se puede calcular la media.")
    else:
        print(f"Media: {media:.2f}")

mostrar_media([3, 5, 8])
mostrar_media([3, "cinco"]) 
```
