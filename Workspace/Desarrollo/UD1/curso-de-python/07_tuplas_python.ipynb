# Tuplas en Python

Una tupla es una secuencia ordenada, como una lista, pero **inmutable**: una vez creada, no se puede modificar.

## Características

- Se escriben habitualmente con paréntesis: `(valor1, valor2)`.
- Conservan el orden y admiten duplicados.
- No permiten `append`, `remove` ni asignación por índice.
- Son apropiadas para datos que no deberían cambiar: coordenadas, fechas o configuraciones.
- Una tupla de un elemento necesita una coma: `(5,)`.

```python
coordenadas = (37.389, -5.984)
punto = (10,)
sin_elementos = ()

print(coordenadas)
print(type(coordenadas))
print(type(punto))
print(type((10)))  # Esto es un int, no una tupla
```

## Acceso y desempaquetado

Las tuplas admiten índices y cortes. También se pueden desempaquetar: asignar sus elementos a varias variables en una sola línea.

```python
rgb = (30, 144, 255)
print(rgb[0])
print(rgb[1:])

rojo, verde, azul = rgb
print(f"R={rojo}, G={verde}, B={azul}")
```

## Inmutabilidad

La inmutabilidad ayuda a evitar cambios accidentales. Una tupla puede contener una lista, pero esa lista interior seguirá siendo mutable.

```python
configuracion = ("modelo-v1", [0.1, 0.2], True)
# configuracion[0] = "modelo-v2"  # Produciría TypeError
configuracion[1].append(0.3)
print(configuracion)

print(configuracion.count(True))
print(configuracion.index("modelo-v1"))
```

## Cuándo elegir tupla o lista

Elige una **tupla** si el conjunto de valores representa una unidad fija, como `(latitud, longitud)`. Elige una **lista** si necesitas añadir, eliminar u ordenar elementos durante el programa.

Las tuplas también pueden utilizarse como claves de diccionarios si sus elementos son inmutables.

```python
ciudad, provincia, pais = ("Sevilla", "Sevilla", "España")
print(f"{ciudad}, {provincia}, {pais}")

# Intercambio elegante de variables
a, b = 3, 7
a, b = b, a
print(a, b)
```
