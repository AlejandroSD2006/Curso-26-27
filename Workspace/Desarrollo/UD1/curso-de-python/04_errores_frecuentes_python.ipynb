# Errores frecuentes en Python

## Cómo leerlos, prevenirlos y corregirlos

Cometer errores es parte normal de programar. La clave es aprender a interpretar el mensaje y localizar su causa.

## Cómo usar este cuaderno

Cada caso incluye:

1. El mensaje de error habitual.
2. La causa.
3. Código con el error, mostrado como texto para no interrumpir el cuaderno.
4. Una solución ejecutable.
5. En muchos casos, una demostración segura con `try` y `except`.

Lee siempre el tipo de error, la línea indicada y el fragmento marcado con `^`.

# 1. Errores de sintaxis

Estos errores aparecen antes de que el programa comience a ejecutarse. Python no puede interpretar el código escrito.

## `SyntaxError: invalid syntax`

**Frecuencia:** muy frecuente

**¿Por qué ocurre?** Python encontró algo que no entiende. Generalmente falta un paréntesis, una comilla o dos puntos.

### Código con error

```python
if edad >= 18   # Falta el : al final
    print("Eres mayor de edad")
```

### Solución

Revisa la línea que marca el error y también la anterior. Busca paréntesis o comillas sin cerrar, y `:` faltantes después de `if`, `for`, `while`, `def`, `class`, `try` o `else`.

````python
edad = 20
if edad >= 18:
    print("Eres mayor de edad")
````

## `IndentationError`

**Frecuencia:** muy frecuente

**¿Por qué ocurre?** Python usa la indentación (los espacios al inicio) para saber qué código pertenece a cada bloque. Si es incorrecta, Python no puede determinar la estructura.

### Código con error

```python
for i in range(5):
print(i)  # Falta indentación
```

Añade 4 espacios al inicio de las líneas dentro de un bloque. No mezcles espacios y tabuladores en el mismo archivo.

````python
for i in range(5):
    print(i)
````

## Cadena sin cerrar

El mensaje puede aparecer como `SyntaxError: EOL while scanning string literal` o, en versiones recientes, `SyntaxError: unterminated string literal`.

**Frecuencia:** frecuente

**¿Por qué ocurre?** Se abrió una comilla sin cerrarla. Python llega al final de línea (`EOL`, *End Of Line*) esperando la comilla de cierre.

### Código con error

```python
mensaje = "Hola, bienvenido al curso  # Falta la comilla final
```

Revisa que las comillas estén en pares. Si el texto contiene comillas, alterna comillas simples y dobles: `"Él dijo 'hola'"` o `'Él dijo "hola"'`. 

````python
mensaje = "Hola, bienvenido al curso"
cita = 'Él dijo "hola"'
print(mensaje)
print(cita)
````

# 2. Errores de variables

Aparecen cuando el programa se ejecuta, pero intenta usar un nombre que no existe o que aún no tiene valor disponible.

## `NameError: name 'x' is not defined`

**Frecuencia:** muy frecuente

**¿Por qué ocurre?** Se usa una variable que no existe: puede tener una errata, no haberse creado todavía o estar fuera del ámbito actual.

### Código con error

```python
nombre = "Juan"
print(nomber)  # Se escribió nomber en vez de nombre
```

Python distingue mayúsculas y minúsculas: `nombre`, `Nombre` y `NOMBRE` son variables diferentes.

````python
nombre = "Juan"
print(nombre)

# Demostración segura
try:
    print(nomber)
except NameError as error:
    print(f"Error detectado: {error}")
````

## `UnboundLocalError`

**Frecuencia:** intermedia

**¿Por qué ocurre?** Dentro de una función se intenta usar una variable que Python considera local, pero todavía no se le ha asignado un valor.

### Código con error

```python
contador = 0

def incrementar():
    contador = contador + 1  # Error: contador se considera local
    return contador
```

Una solución es declarar `global contador`, pero normalmente es mejor pasar el valor como parámetro y devolver el resultado.

````python
def incrementar(contador):
    return contador + 1

contador = 0
contador = incrementar(contador)
print(contador)
````

# 3. Errores de tipo

Un tipo determina qué operaciones admite un valor. Comprueba los tipos con `type()` o `isinstance()` cuando tengas dudas.

## `TypeError: can only concatenate str to str`

**Frecuencia:** muy frecuente

**¿Por qué ocurre?** Se intenta concatenar un texto y un número. Python no convierte automáticamente entre esos tipos.

### Código con error

```python
edad = 25
mensaje = "Tengo " + edad + " años"
```

Convierte el número con `str(edad)` o, preferiblemente, usa una *f-string*.

````python
edad = 25
mensaje = f"Tengo {edad} años"
print(mensaje)
print(type(edad), type(mensaje))
````

## `TypeError: 'int' object is not subscriptable`

**Frecuencia:** frecuente

**¿Por qué ocurre?** Se usan corchetes `[]` sobre un valor que no admite índices, como un entero. Las listas, cadenas, tuplas y diccionarios sí permiten acceso con corchetes, cada uno a su manera.

### Código con error

```python
numero = 12345
primer_digito = numero[0]
```

Si se necesita acceder a los dígitos, convierte primero el número en texto: `str(numero)[0]`.

````python
numero = 12345
primer_digito = str(numero)[0]
print(primer_digito)
print(type(numero), type(str(numero)))
````

# 4. Errores de índice y clave

Las colecciones tienen límites y reglas de acceso. Comprueba lo que contienen antes de pedir un elemento.

## `IndexError: list index out of range`

**Frecuencia:** muy frecuente

**¿Por qué ocurre?** Se intenta acceder a una posición que no existe. Los índices empiezan en 0.

### Código con error

```python
frutas = ["manzana", "banana", "naranja"]  # Índices: 0, 1, 2
print(frutas[3])
```

Usa `len(lista)` para conocer el número de elementos o `lista[-1]` para acceder al último.

````python
frutas = ["manzana", "banana", "naranja"]
print(f"Hay {len(frutas)} frutas")
print(f"La última es {frutas[-1]}")

try:
    print(frutas[3])
except IndexError as error:
    print(f"Error detectado: {error}")
````

## `KeyError`

**Frecuencia:** frecuente

**¿Por qué ocurre?** Se accede mediante `diccionario[clave]` a una clave que no existe.

### Código con error

```python
persona = {"nombre": "Ana", "edad": 30}
print(persona["apellido"])
```

Usa `.get()` si la clave puede no existir. Permite indicar un valor por defecto.

````python
persona = {"nombre": "Ana", "edad": 30}
apellido = persona.get("apellido", "No especificado")
print(apellido)

try:
    print(persona["apellido"])
except KeyError as error:
    print(f"Error detectado: {error}")
````

# 5. Errores con `None`

`None` representa la ausencia de valor. Muchas funciones modifican un objeto pero no devuelven ningún resultado; por eso su retorno es `None`.

## `AttributeError: 'NoneType' has no attribute ...`

**Frecuencia:** frecuente

**¿Por qué ocurre?** La variable vale `None` cuando se esperaba una lista, texto u otro objeto. Un caso habitual es asignar el resultado de un método que modifica una lista *in situ*, como `.sort()`.

### Código con error

```python
lista = [3, 1, 2]
lista_ordenada = lista.sort()  # sort() devuelve None
print(lista_ordenada[0])
```

`.sort()` modifica la lista original. Para obtener una lista nueva, utiliza `sorted(lista)`.

````python
lista = [3, 1, 2]
lista_ordenada = sorted(lista)
print(lista_ordenada[0])
print(f"Original: {lista}; nueva: {lista_ordenada}")

otra_lista = [3, 1, 2]
resultado = otra_lista.sort()
print(f"El retorno de .sort() es {resultado!r}")
````

# 6. Estrategia para depurar

1. Lee el tipo de error: indica la categoría del problema.
2. Localiza la línea indicada, pero revisa también las anteriores.
3. Muestra valores y tipos con `print(valor, type(valor))`.
4. Reduce el problema a un ejemplo pequeño.
5. Consulta la documentación con `help()` cuando no conozcas una función o método.
6. Corrige una cosa cada vez y vuelve a ejecutar.

## Mini-reto de depuración

Corrige este programa para que muestre la primera letra del nombre y el último valor de la lista. Identifica al menos dos problemas antes de modificarlo.

```python
nombre = None
puntuaciones = [8, 9, 7]
print(nombre[0])
```

# Ideas clave

- Los mensajes de error son información útil, no un fracaso.
- `SyntaxError` e `IndentationError` impiden iniciar el programa.
- `NameError`, `TypeError`, `IndexError`, `KeyError` y `AttributeError` aparecen durante la ejecución.
- Conocer los tipos, validar entradas y usar funciones adecuadas previene muchos errores.

La práctica de leer, reproducir y corregir errores es una habilidad central de programación.
