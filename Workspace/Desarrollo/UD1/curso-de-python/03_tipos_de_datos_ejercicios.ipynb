# Bloque D — Usar el tipo correctamente

## D1. Clasificador de valores

Completa la función para que devuelva `"número"`, `"texto"`, `"colección"` u `"otro"`. Usa `isinstance()`, no comparaciones con el resultado de `type()`.

```python
def categoria(valor):
    if isinstance(valor, (int, float)) and not isinstance(valor, bool):
        return "número"
    if isinstance(valor, str):
        return "texto"
    if isinstance(valor, (list, tuple, set, dict)):
        return "colección"
    return "otro"

for ejemplo in [4, 2.7, True, "IA", [1, 2], None]:
    print(f"{ejemplo!r:10} -> {categoria(ejemplo)}")
```

## D2. Función reutilizable: media

La función `media` introduce una idea muy usada: recibir datos, validarlos y devolver un resultado. Prueba con listas de enteros, decimales, una lista vacía y una lista que incluya texto.

```python
def media(valores):
    """Devuelve la media de una lista de números o None si no es válida."""
    if not isinstance(valores, list) or not valores:
        return None
    if not all(isinstance(x, (int, float)) and not isinstance(x, bool) for x in valores):
        return None
    return sum(valores) / len(valores)

casos = [[6, 8, 10], [1.5, 2.5], [], [1, "dos"], [True, 4]]
for caso in casos:
    print(f"{caso!r:15} -> {media(caso)}")
```

## D3. Conversión robusta

Completa `a_float` para devolver un número decimal o `None` si no puede convertir el valor. Prueba con `"3.5"`, `7`, `None` y `"tres"`.

Pista: una conversión puede fallar con `ValueError` o `TypeError`.

```python
def a_float(valor):
    try:
        return float(valor)
    except (ValueError, TypeError):
        return None

for dato in ["3.5", 7, None, "tres"]:
    print(f"{dato!r:8} -> {a_float(dato)}")
```

## E1. Limpieza de datos de un experimento

`mediciones` mezcla textos numéricos, números válidos, un dato perdido y un texto incorrecto.

1. Convierte a `float` los valores que sea posible.
2. Descarta los que no se puedan convertir.
3. Muestra los valores válidos, su tipo y su media.
4. ¿Por qué una comprobación de tipo sola no sería suficiente en este caso?

```python
mediciones = ["18.5", 20, None, "error", 19.2, "21"]

validas = []
for medicion in mediciones:
    try:
        validas.append(float(medicion))
    except (ValueError, TypeError):
        pass

print(validas)
print([type(x).__name__ for x in validas])
print(f"Media: {sum(validas) / len(validas):.2f}")
```
