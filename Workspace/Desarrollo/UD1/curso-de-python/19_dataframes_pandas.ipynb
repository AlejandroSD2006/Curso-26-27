# DataFrames en Pandas

Un `DataFrame` es una tabla bidimensional con filas y columnas etiquetadas. Es la estructura central de Pandas para análisis de datos tabulares.

## Crear DataFrames

Podemos crearlos desde diccionarios de listas, listas de diccionarios, listas de listas o archivos como CSV. Cada columna debe tener la misma cantidad de valores.

```python
import pandas as pd

alumnado = pd.DataFrame({
    "nombre": ["Ada", "Lin", "Noa"],
    "edad": [20, 22, 21],
    "nota": [8.5, 7.0, 9.2]
})
display(alumnado)
```

## Inspección inicial

- `.shape`: número de filas y columnas.
- `.columns`: nombres de columnas.
- `.dtypes`: tipos de datos.
- `.head(n)` y `.tail(n)`: primeras o últimas filas.
- `.info()`: resumen técnico.
- `.describe()`: resumen estadístico de columnas numéricas.

```python
print(alumnado.shape)
print(alumnado.columns.tolist())
print(alumnado.dtypes)
display(alumnado.head(2))
display(alumnado.describe())
```

## Seleccionar columnas y filas

- `df["columna"]` devuelve una `Series`.
- `df[["a", "b"]]` devuelve un `DataFrame`.
- `.loc[fila, columna]` usa etiquetas.
- `.iloc[fila, columna]` usa posiciones enteras.

```python
print(alumnado["nombre"])
display(alumnado[["nombre", "nota"]])
print(alumnado.loc[1, "nombre"])
print(alumnado.iloc[1, 2])
display(alumnado.loc[0:1, ["nombre", "edad"]])
```

## Añadir y modificar columnas

Una operación entre Series se realiza por fila. Podemos crear columnas calculadas o asignar valores nuevos. Para modificar parte de una tabla, usa `.loc`.

```python
alumnado["aprobado"] = alumnado["nota"] >= 5
alumnado["nota_sobre_10"] = alumnado["nota"]
alumnado.loc[alumnado["nombre"] == "Lin", "edad"] = 23
display(alumnado)
```

## Índice

Por defecto el índice es 0, 1, 2... Podemos usar una columna como índice si contiene identificadores únicos. El índice facilita la selección por etiqueta.

```python
por_nombre = alumnado.set_index("nombre")
display(por_nombre)
print(por_nombre.loc["Ada", "nota"])

# reset_index() devuelve el índice a una columna
display(por_nombre.reset_index())
```
