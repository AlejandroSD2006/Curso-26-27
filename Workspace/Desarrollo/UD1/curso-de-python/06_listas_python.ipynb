# Listas en Python

Una lista guarda una secuencia ordenada de valores. Es uno de los tipos más utilizados para trabajar con conjuntos de datos.

## Características

- Se crean con corchetes: `[]`.
- Conservan el orden de inserción.
- Admiten valores repetidos y de tipos distintos.
- Son **mutables**: podemos añadir, modificar o eliminar elementos.
- Sus índices comienzan en `0`.

```python
modelos = ["regresión", "árbol", "red neuronal"]
medidas = [0.82, 0.91, 0.88]
mixta = ["modelo A", 0.91, True]

print(modelos)
print(type(modelos))
print(len(modelos))
```

## Acceso, cortes y modificación

Usamos un índice para obtener o modificar una posición. También podemos extraer partes de una lista mediante cortes: `lista[inicio:fin]`.

```python
frutas = ["manzana", "banana", "naranja", "pera"]
print(frutas[0])
print(frutas[-1])
print(frutas[1:3])

frutas[1] = "plátano"
print(frutas)
```

## Métodos frecuentes

- `.append(x)`: añade al final.
- `.insert(i, x)`: inserta en una posición.
- `.remove(x)`: elimina la primera aparición de un valor.
- `.pop()`: extrae y devuelve un elemento.
- `.sort()`: ordena la lista original.
- `.count(x)`: cuenta apariciones.

```python
notas = [7, 9, 6, 9]
notas.append(8)
notas.insert(0, 10)
notas.remove(6)
ultima = notas.pop()
notas.sort()
print(notas)
print(f"Elemento extraído: {ultima}; número de 9: {notas.count(9)}")
```

## Recorrer una lista

El bucle `for` procesa cada elemento. `enumerate()` aporta también su índice; es preferible a gestionar el contador manualmente.

```python
sensores = ["temperatura", "humedad", "luz"]
for posicion, sensor in enumerate(sensores, start=1):
    print(f"{posicion}. {sensor.title()}")
```
