# Agregación de Series en Pandas

Agregar es resumir muchos valores mediante una medida: suma, media, mínimo, máximo, cantidad o frecuencia. Es una fase esencial en el análisis de datos.

## Agregaciones numéricas básicas

Las Series numéricas incorporan métodos como `.sum()`, `.mean()`, `.median()`, `.min()`, `.max()`, `.count()`, `.std()` y `.var()`. Por defecto, muchos ignoran valores ausentes.

```python
import pandas as pd

ventas = pd.Series([120, 150, None, 100, 180], name="ventas")
print(f"Suma: {ventas.sum()}")
print(f"Media: {ventas.mean():.2f}")
print(f"Mediana: {ventas.median():.2f}")
print(f"Mínimo: {ventas.min()}; máximo: {ventas.max()}")
print(f"Datos no nulos: {ventas.count()}; total: {ventas.size}")
```

## Posición de extremos

`.idxmin()` y `.idxmax()` devuelven las etiquetas de índice del mínimo y máximo; `.min()` y `.max()` devuelven los valores. Es una diferencia importante.

```python
notas = pd.Series([8.5, 7.0, 9.2], index=["Ada", "Lin", "Noa"])
print(notas.max(), notas.idxmax())
print(notas.min(), notas.idxmin())
print(f"Mejor nota: {notas.idxmax()} ({notas.max()})")
```

## Resumen con `describe()` y `agg()`

`.describe()` genera un resumen estadístico. `.agg()` permite pedir varias agregaciones en una sola operación, incluso usando nombres de métodos o funciones propias.

```python
print(ventas.describe())
print(ventas.agg(["count", "sum", "mean", "min", "max"]))

resumen = notas.agg(media="mean", mejor="max", peor="min")
print(resumen)
```

## Frecuencias en Series de texto

`.value_counts()` cuenta apariciones de cada valor. Con `normalize=True` devuelve proporciones. Es una agregación muy útil para categorías.

```python
etiquetas = pd.Series(["gato", "perro", "gato", "ave", "gato", "perro"])
print(etiquetas.value_counts())
print(etiquetas.value_counts(normalize=True))
print(f"Categorías distintas: {etiquetas.nunique()}")
```

## Agregaciones acumuladas

`.cumsum()` calcula sumas acumuladas; `.cummax()` conserva el máximo visto hasta cada posición. Son útiles para series temporales y seguimiento de objetivos.

```python
diarias = pd.Series([15, 20, 12, 25], index=["L", "M", "X", "J"])
print(diarias.cumsum())
print(diarias.cummax())
```
