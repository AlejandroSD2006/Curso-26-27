# Pandas: bibliotecas e importación

Pandas es una biblioteca de Python para trabajar con datos tabulares y series de datos. Aporta las estructuras `Series` y `DataFrame`.

## ¿Qué es una biblioteca?

Una biblioteca reúne código creado por otras personas que podemos reutilizar. Antes de importarla debe estar instalada en el entorno de Python.

En una terminal se instala con:

```bash
python -m pip install pandas
```

En una celda de cuaderno también puede usarse `%pip install pandas`.

## Importar Pandas

La convención universal es importar Pandas como `pd`. El alias acorta el código y hace que otros lectores reconozcan inmediatamente la biblioteca utilizada.

```python
import pandas as pd
```

````python
import pandas as pd

print(pd.__version__)
print(pd.Series([10, 20, 30]))
````

## Importar objetos concretos

También podemos importar una clase o función concreta. Es útil si se emplea mucho, aunque para principiantes suele ser más claro mantener el prefijo `pd.`.

```python
from pandas import Series
from pandas import DataFrame as DF
```

Renombrar con `as` crea un alias local; no cambia la biblioteca.

````python
from pandas import Series
from pandas import DataFrame as DF

temperaturas = Series([18.5, 19.1, 20.0])
tabla = DF({"ciudad": ["Sevilla", "Córdoba"], "maxima": [31, 29]})
print(temperaturas)
display(tabla)
````

## Importaciones recomendadas

En análisis de datos es habitual combinar Pandas con otras bibliotecas:

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

Cada alias es una convención común. Evita `from pandas import *`: puede ocultar el origen de los nombres y provocar conflictos.

````python
# Ejemplo de inspección inicial
datos = pd.Series(["Python", "Pandas", "IA"])
print(type(datos))
print(datos.dtype)
print([nombre for nombre in dir(pd) if nombre in ["Series", "DataFrame", "read_csv"]])
````
