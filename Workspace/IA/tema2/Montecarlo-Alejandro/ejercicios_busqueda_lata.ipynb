## Optimización de una lata de refresco

El objetivo es estimar las dimensiones de una lata cilíndrica de **330 ml** que minimicen la cantidad de chapa de aluminio. Como $1\,\mathrm{ml}=1\,\mathrm{cm}^3$, el volumen constante es $V=330\,\mathrm{cm}^3$.

Para un radio $r$ y una altura $h$, las fórmulas son:

$$V=\pi r^2h, \qquad A=2\pi r^2+2\pi rh$$

El área incluye las dos tapas y el lateral del cilindro.

## Ejercicio 1: búsqueda Montecarlo básica

**Enunciado.** Utiliza el método de simulación de Montecarlo (ruleta) para determinar las dimensiones de una lata de refrescos de 330 ml que minimicen la cantidad de chapa de aluminio. Modela la lata como un cilindro con radio y altura. En cada iteración, elige un radio aleatorio entre 1,5 y 8 cm, calcula el área de las dos tapas y el lateral, y guarda el área mínima y el radio que la produjo. Aunque el problema puede resolverse derivando, se pide utilizar Montecarlo. Realiza 10.000 iteraciones.

```python
import math
import random

volumen_objetivo = 330.0  # cm^3
numero_iteraciones = 10_000
radio_minimo = 1.5  # cm
radio_maximo = 8.0  # cm
random.seed(42)

mejor_area = float('inf')
mejor_radio = 0.0
mejor_altura = 0.0

for _ in range(numero_iteraciones):
    radio = random.uniform(radio_minimo, radio_maximo)
    altura = volumen_objetivo / (math.pi * radio**2)
    area = 2 * math.pi * radio**2 + 2 * math.pi * radio * altura
    if area < mejor_area:
        mejor_area = area
        mejor_radio = radio
        mejor_altura = altura

print(f'Radio encontrado: {mejor_radio:.4f} cm')
print(f'Altura correspondiente: {mejor_altura:.4f} cm')
print(f'Área mínima encontrada: {mejor_area:.4f} cm^2')
```

## Ejercicio 2: mostrar cada mejora

**Enunciado.** Modifica el algoritmo anterior para que cada vez que encuentre una solución mejor, la muestre por pantalla.

```python
import math
import random

volumen_objetivo = 330.0  # cm^3
numero_iteraciones = 10_000
radio_minimo = 1.5  # cm
radio_maximo = 8.0  # cm
random.seed(42)

mejor_area = float('inf')
mejor_radio = 0.0
mejor_altura = 0.0

for iteracion in range(1, numero_iteraciones + 1):
    radio = random.uniform(radio_minimo, radio_maximo)
    altura = volumen_objetivo / (math.pi * radio**2)
    area = 2 * math.pi * radio**2 + 2 * math.pi * radio * altura
    if area < mejor_area:
        mejor_area = area
        mejor_radio = radio
        mejor_altura = altura
        print(f'Iteración {iteracion}: nueva mejor solución: radio = {mejor_radio:.4f} cm, altura = {mejor_altura:.4f} cm, área = {mejor_area:.4f} cm^2')

print('\nMejor solución al finalizar:')
print(f'Radio: {mejor_radio:.4f} cm')
print(f'Altura: {mejor_altura:.4f} cm')
print(f'Área mínima: {mejor_area:.4f} cm^2')
```

## Ejercicio 3: búsqueda en dos fases

**Enunciado.** Modifica el algoritmo anterior para que la búsqueda se produzca en dos fases. La primera fase realizará 2.000 lanzamientos de ruleta. Con el mejor radio obtenido, ajusta los límites a $r_{min}=r_{mejor}-1,50$ y $r_{max}=r_{mejor}+1,50$. Con estos nuevos límites, realiza 8.000 iteraciones y muestra por pantalla el valor obtenido. ¿Qué algoritmo se está emulando con esta variante? Justifica tu respuesta.

La mejor área y sus dimensiones se conservan entre ambas fases.

```python
import math
import random

volumen_objetivo = 330.0  # cm^3
radio_minimo = 1.5  # cm
radio_maximo = 8.0  # cm
random.seed(42)

mejor_area = float('inf')
mejor_radio = 0.0
mejor_altura = 0.0

# Fase 1: exploración del intervalo inicial.
for iteracion in range(1, 2_000 + 1):
    radio = random.uniform(radio_minimo, radio_maximo)
    altura = volumen_objetivo / (math.pi * radio**2)
    area = 2 * math.pi * radio**2 + 2 * math.pi * radio * altura
    if area < mejor_area:
        mejor_area = area
        mejor_radio = radio
        mejor_altura = altura
        print(f'Fase 1, iteración {iteracion}: nueva mejor solución: radio = {mejor_radio:.4f} cm, área = {mejor_area:.4f} cm^2')

# Fase 2: refinamiento alrededor del mejor radio de la fase 1.
radio_minimo = mejor_radio - 1.50
radio_maximo = mejor_radio + 1.50
print(f'\nIntervalo de la fase 2: [{radio_minimo:.4f}, {radio_maximo:.4f}] cm')

for iteracion in range(1, 8_000 + 1):
    radio = random.uniform(radio_minimo, radio_maximo)
    altura = volumen_objetivo / (math.pi * radio**2)
    area = 2 * math.pi * radio**2 + 2 * math.pi * radio * altura
    if area < mejor_area:
        mejor_area = area
        mejor_radio = radio
        mejor_altura = altura
        print(f'Fase 2, iteración {iteracion}: nueva mejor solución: radio = {mejor_radio:.4f} cm, área = {mejor_area:.4f} cm^2')

print('\nMejor solución final:')
print(f'Radio: {mejor_radio:.4f} cm')
print(f'Altura: {mejor_altura:.4f} cm')
print(f'Área mínima: {mejor_area:.4f} cm^2')
```

### Respuesta: algoritmo emulado

Esta variante emula un **algoritmo de mejora iterativa**: prueba soluciones repetidamente y conserva la mejor encontrada, actualizándola solo cuando aparece otra con menor área. Por eso, el valor de la función objetivo mejora o se mantiene a lo largo de la búsqueda.

La búsqueda se hace en dos fases: primero explora todo el intervalo de radios y después realiza nuevos intentos en un intervalo más pequeño, centrado en el mejor radio encontrado. Así refina progresivamente la solución mediante pruebas aleatorias.
